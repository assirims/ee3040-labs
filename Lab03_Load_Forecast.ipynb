# Part 1 · Forecast scores, by hand

Every forecast is judged by its **misses**: real value minus forecast.
Three numbers summarise them, and each answers a different question.

| Number | Formula | The question it answers |
|---|---|---|
| **MAE** mean absolute error | average of \|miss\| | "On average, how far off are we, in points of loading?" The number the control-room engineer feels. |
| **RMSE** root mean squared error | √(average of miss²) | Same units, but big misses count much more. The number for "did we miss the peak?" |
| **R²** | 1 − Σmiss² / Σ(real − mean)² | "What share of the ups and downs did we explain?" 1 is perfect, 0 is no better than always predicting the average. |

A warm-up on four validation hours (the same four as in the lecture), loading in % of rating:

| real (%) | forecast (%) |
|---|---|
| 30 | 28 |
| 55 | 52 |
| 75 | 76 |
| 40 | 45 |

### ✏️ Your turn 1 · The three scores
Complete the three lines.
NumPy works on whole arrays at once: `miss ** 2` squares every miss, `np.abs(miss)` takes every absolute value, and `.mean()` and `.sum()` do what they say.
<details><summary>💡 Hint</summary>

MAE: take the absolute value of every miss, then the average.
RMSE: square every miss, average the squares, then take the square root with `np.sqrt`.
R²: follow the formula in the table; `real.mean()` is the average of the real values.
</details>
<details><summary>🔑 Still stuck? The code</summary>

`mae = np.abs(miss).mean()`, `rmse = np.sqrt((miss ** 2).mean())`, `r2 = 1 - (miss ** 2).sum() / ((real - real.mean()) ** 2).sum()`.
</details>

RMSE is never smaller than MAE, and the two are equal only when every miss has the same size.
**How far** RMSE sits above MAE shows how uneven the misses are. Here the one 5-point miss opens the gap, because squaring makes big misses loud.
When RMSE is much larger than MAE, a few big misses are hiding behind a good average. Keep that in mind; it is the story of this lab.

And R²? Always forecasting the average of the four (50%) would miss by an RMSE of about 17 points.
Your forecast's RMSE is a small fraction of that, which is exactly what an R² close to 1 says.

**الجزء 2 · الجدول النظيف وخط أساس 2024**

🟢 شغّل الخلية المخفية. تعيد بناء الجدول الساعي النظيف من اللاب 2 (⁦`hourly`⁩) بخطوات التنظيف نفسها التي كتبتها.
وتجمع دالة مرجعية واحدة تلك الخطوات، حتى يبدأ كل لاب من المكان نفسه.

```python
#@title 🟢 Load the course data (run this cell; no need to read it)
import numpy as np
import pandas as pd

SEED = 3040
OVERLOAD_PCT = 100.0
HOT_MONTHS = [5, 6, 7, 8, 9]


def _temperature(hours, rng):
    """Hourly air temperature in °C for a hot, dry city: a seasonal wave, a daily cycle that
    peaks mid-afternoon, day-to-day weather that persists, and a few summer heat waves."""
    doy = hours.dayofyear.values
    hour = hours.hour.values
    season = 27.0 + 11.0 * np.sin(2 * np.pi * (doy - 110) / 365.25)
    daily = 6.0 * np.cos(2 * np.pi * (hour - 15) / 24)
    days = (hours.normalize() - hours[0].normalize()).days.values
    n_days = days.max() + 1
    anom = np.zeros(n_days)
    for d in range(1, n_days):                                  # weather persists for days
        anom[d] = 0.8 * anom[d - 1] + rng.normal(0, 1.0)
    day_ts = pd.date_range(hours[0].normalize(), periods=n_days, freq="D")
    summer = np.isin(day_ts.month, [6, 7, 8])
    for s in np.flatnonzero(summer & (rng.random(n_days) < 2.2 / 92)):   # about two heat waves a summer
        anom[s:s + int(rng.integers(3, 8))] += rng.uniform(2.0, 4.0)
    t = season + daily + anom[days] + rng.normal(0, 0.5, len(hours))
    return np.round(t, 1)


def _tie(hours, rng):
    """Tie switch to the neighbouring feeder: closed (1) except during maintenance windows,
    about ten a year, 6 to 30 hours each, when this transformer carries the whole load."""
    n = len(hours)
    closed = np.ones(n, dtype=int)
    for s in np.flatnonzero(rng.random(n) < 10 / 8760):
        closed[s:s + int(rng.integers(6, 31))] = 0
    return closed


# residential load shape by hour, as a fraction of the daily peak (evening peak, quiet dawn)
_PROFILE = np.array([.62, .56, .52, .49, .47, .47, .50, .55, .60, .64, .67, .70,
                     .72, .74, .76, .78, .80, .83, .88, .93, .96, .94, .85, .73])

TRANSFORMER = dict(base=42.0, ac=1.15, ac_start=24.0, ac_power=1.45, share=0.84, growth=0.02)


def _ramadan(hours):
    """Approximate Ramadan months for 2021-2025 (the evening peak moves later and grows)."""
    starts = {2021: "2021-04-13", 2022: "2022-04-02", 2023: "2023-03-23", 2024: "2024-03-11", 2025: "2025-03-01"}
    r = np.zeros(len(hours), dtype=bool)
    for y, s in starts.items():
        s = pd.Timestamp(s)
        r |= (hours >= s) & (hours < s + np.timedelta64(30, "D"))
    return r


def _simulate(hours, temp, tie, rng, p=TRANSFORMER):
    """Loading in % of rating, and the neighbouring feeder's loading."""
    n = len(hours)
    t = pd.Series(temp)
    felt = (0.45 * t + 0.55 * t.rolling(24, min_periods=1).mean()).values   # buildings store heat
    ac = p["ac"] * np.clip(felt - p["ac_start"], 0, None) ** p["ac_power"]
    shape = _PROFILE[hours.hour.values]
    weekend = np.isin(hours.dayofweek.values, [4, 5])                        # Friday and Saturday
    ram = _ramadan(hours)
    evening = np.isin(hours.hour.values, [20, 21, 22, 23, 0, 1])
    base = p["base"] * shape * np.where(weekend, 1.06, 1.0) * np.where(ram & evening, 1.15, 1.0)
    noise = np.zeros(n)
    for i in range(1, n):                                                    # slow, correlated noise
        noise[i] = 0.9 * noise[i - 1] + rng.normal(0, 1.1)
    years = (hours.year.values - 2021) + hours.dayofyear.values / 365.0
    demand = (base + ac * (0.55 + 0.45 * shape) + noise) * (1 + p.get("growth", 0.02)) ** years   # new customers every year
    loading = demand * np.where(tie == 1, p["share"], 1.0)
    neighbour = np.roll(0.9 * demand, 1) + rng.normal(0, 2.0, n)              # the neighbouring feeder follows the same heat, an hour later
    return np.round(np.clip(neighbour, 0, None), 1), np.round(np.clip(loading, 0, None), 1)


def _overload_events(loading):
    above = loading > OVERLOAD_PCT
    return int(np.sum(above[1:] & ~above[:-1]) + (1 if above[0] else 0))


def _overload_report_column(df):
    """Filled only on days the transformer overloaded: the time the control room logged the overload."""
    over = df["loading_pct"] > OVERLOAD_PCT
    first = df.loc[over].groupby(df.loc[over, "timestamp"].dt.date)["timestamp"].min()
    report = pd.Series("", index=df.index, dtype=object)
    day = df["timestamp"].dt.date
    for d, ts in first.items():
        report[day == d] = (ts + np.timedelta64(40, "m")).strftime("%H:%M")
    return report


def _add_mess(df, rng, gap=True):
    df = df.copy()
    n = len(df)
    val = df["loading_pct"].values.copy()
    # 1. bursts of the logger's 'no reading' code
    for s in rng.choice(n - 12, size=int(round(n * 0.0328 / 6)), replace=False):
        val[s:s + int(rng.integers(3, 10))] = -999.0
    # 2. dropouts to exactly zero
    for s in rng.choice(n - 14, size=max(2, int(round(n * 0.0092 / 9))), replace=False):
        val[s:s + int(rng.integers(6, 13))] = 0.0
    df["loading_pct"] = val
    # 3. three temperatures logged in tenths of a degree (history only)
    if gap:
        hot = np.flatnonzero((df["temp_c"].values > 45) & (df["temp_c"].values < 49))
        rows = rng.choice(hot, size=3, replace=False)
        df.loc[rows, "temp_c"] = (df.loc[rows, "temp_c"] * 10).round(0)
    # 4. the logger replacement: an 11-day gap, then a CT that reads 4% high (history only)
    if gap:
        cut0, cut1 = pd.Timestamp("2022-11-08"), pd.Timestamp("2022-11-19")
        df = df[(df["timestamp"] < cut0) | (df["timestamp"] >= cut1)].copy()
    after = df["timestamp"] >= pd.Timestamp("2022-11-19")
    real = df["loading_pct"] > 0
    df.loc[after & real, "loading_pct"] = (df.loc[after & real, "loading_pct"] * 1.04).round(1)
    return df.reset_index(drop=True)


def make_transformer(seed=SEED):
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2021-01-01", "2025-12-31 23:00", freq="h")
    temp = _temperature(hours, rng)
    tie = _tie(hours, rng)
    neighbour, loading = _simulate(hours, temp, tie, rng)
    df = pd.DataFrame({"timestamp": hours, "temp_c": temp, "neighbour_load_pct": neighbour,
                       "tie_closed": tie, "loading_pct": loading})
    df["temp_3h_c"] = df["temp_c"].rolling(3, min_periods=1).mean().round(1)
    df["overload_report_time"] = _overload_report_column(df)
    df = df[["timestamp", "temp_c", "temp_3h_c", "neighbour_load_pct", "tie_closed", "loading_pct", "overload_report_time"]]
    hist = df[df["timestamp"] < "2025-01-01"].reset_index(drop=True)
    test = df[df["timestamp"] >= "2025-01-01"].reset_index(drop=True)
    return _add_mess(hist, rng, gap=True), _add_mess(test, rng, gap=False)


# ---------------------------------------------------------------------------
# Track D · the asset register (Lab 5). One row per transformer; answer = faults next summer.
def make_assets(seed=SEED + 1, n=900):
    """900 distribution transformers with their asset-register columns and the faults they had next summer."""
    rng = np.random.default_rng(seed)
    district = rng.choice(["North", "South", "East"], size=n, p=[0.45, 0.30, 0.25])
    age = rng.integers(1, 41, n)
    rating = rng.choice([500, 1000, 1500], size=n, p=[0.30, 0.50, 0.20])
    peak = np.round(np.clip(rng.normal(78, 14, n), 35, 130), 1)
    customers = np.round(rating * rng.uniform(0.18, 0.32, n)).astype(int)
    cable_age = rng.integers(1, 36, n)
    acetylene = np.round(rng.gamma(1.2, 1.6, n) + 6 * (rng.random(n) < 0.07), 1)   # a gas from arcing inside the oil
    fans = (rng.random(n) < 0.35).astype(int)                                   # forced cooling fitted
    # Faults come from thresholds and interactions (overloaded and old, arcing gas, old cables under
    # heavy loading, no fans on the busiest units): easy for trees, hard for a straight line.
    lam = np.exp(-0.2 + 1.1 * (1.0 * ((peak > 95) & (age > 20)) + 0.9 * (acetylene > 5)
                               + 0.7 * (age > 30) + 0.5 * ((cable_age > 25) & (peak > 85))
                               + 0.4 * ((fans == 0) & (peak > 100)))
                 + 0.15 * np.log1p(customers / 50) + 0.01 * age + 0.1 * (district == "South"))
    faults = rng.poisson(lam)
    return pd.DataFrame({
        "transformer_id": [f"T{i:03d}" for i in range(1, n + 1)],
        "district": district,
        "age_years": age,
        "rating_kva": rating,
        "peak_loading_pct": peak,
        "customers": customers,
        "cable_age_years": cable_age,
        "acetylene_ppm": acetylene,
        "cooling_fans": fans,
        "faults_last_3_summers": rng.poisson(1.5 * np.sqrt(lam) * rng.uniform(0.3, 1.7, n)),   # history helps, noisily
        "survey_batch": rng.integers(1, 60, n),
        "crew_hours_next_summer": np.round(np.clip(faults * 11.5 + rng.normal(0, 3, n), 0, None)),
        "faults_next_summer": faults,
    })


# ---------------------------------------------------------------------------
# Track C · forty transformer load loggers (Lab 6). Hourly loading for 2024, no labels.
LOGGER_ZONES = ["residential"] * 10 + ["commercial"] * 14 + ["government"] * 10 + ["farms"] * 6

_ZONE_SHAPE = {   # loading by hour as a fraction of the zone's daily peak
    "residential": _PROFILE,
    "commercial": np.array([.25, .20, .18, .17, .17, .18, .22, .35, .60, .80, .90, .95,
                            .95, .92, .90, .93, .97, 1.0, 1.0, .98, .92, .80, .55, .35]),
    "government": np.array([.20, .18, .17, .17, .17, .20, .40, .80, .95, 1.0, 1.0, 1.0,
                            .98, .95, .70, .40, .30, .27, .25, .24, .23, .22, .21, .20]),
    "farms": np.array([.70, .72, .74, .75, .75, .72, .60, .50, .48, .50, .55, .60,
                       .65, .66, .66, .64, .60, .58, .60, .64, .66, .68, .69, .70]),
}
_ZONE_PEAK = {"residential": 85.0, "commercial": 80.0, "government": 70.0, "farms": 60.0}


def make_loggers(seed=SEED + 2):
    """Hourly loading of 40 transformer loggers in 2024 (long format), and the maintenance log
    the operator 'forgot to mention'."""
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2024-01-01", "2024-12-31 23:00", freq="h")
    n = len(hours)
    temp = _temperature(hours, rng)
    felt = pd.Series(temp).rolling(24, min_periods=1).mean().values
    heat = np.clip(felt - 22.0, 0, None) / 16.0
    frames = []
    for k in range(40):
        lid, zone = f"L{k + 1:02d}", LOGGER_ZONES[k]
        shape = _ZONE_SHAPE[zone][hours.hour.values]
        level = _ZONE_PEAK[zone] * rng.uniform(0.9, 1.1)
        load = level * shape * (0.45 + 0.55 * heat) + rng.normal(0, 1.2, n)
        frames.append(pd.DataFrame({"timestamp": hours, "logger_id": lid, "zone": zone, "loading_pct": load}))
    df = pd.concat(frames, ignore_index=True)
    # a few loggers get a problem each: finding them is the point of Lab 6
    stuck = (df.logger_id == "L17") & (df.timestamp >= "2024-02-03") & (df.timestamp < "2024-02-29")
    df.loc[stuck, "loading_pct"] = rng.normal(0.0, 0.2, int(stuck.sum()))
    df.loc[(df.logger_id == "L31") & (df.timestamp >= "2024-11-12"), "loading_pct"] += 4.0
    days = np.sort(rng.choice(np.arange(5, 360), size=14, replace=False))
    log = []
    for d in days:
        start = pd.Timestamp("2024-01-01") + np.timedelta64(int(d) * 24 + int(rng.integers(8, 12)), "h")
        end = start + np.timedelta64(int(rng.integers(4, 9)), "h")
        w = (df.logger_id == "L08") & (df.timestamp >= start) & (df.timestamp < end)
        df.loc[w, "loading_pct"] = np.abs(rng.normal(0.0, 0.3, int(w.sum())))
        log.append({"logger_id": "L08", "start": start, "end": end, "work": "switchgear maintenance, transformer de-energized"})
    df["loading_pct"] = df["loading_pct"].clip(lower=0).round(1)
    return df, pd.DataFrame(log)


# ---------------------------------------------------------------------------
# Twelve transformers in two districts (Lab 9): who does the model fail for?
SITES = ([(f"N{i}", "North", 2016 + i % 3) for i in range(1, 9)]
         + [(f"S{i}", "South", 2022) for i in range(1, 5)])


def make_network(seed=SEED + 3):
    """Hourly rows for 12 transformers, ready to model. Columns: site, district, logger_since,
    timestamp, the model inputs, and overload_in_6h (1 if the true loading passes 100% within the
    next six hours)."""
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2021-01-01", "2025-12-31 23:00", freq="h")
    city_temp = _temperature(hours, rng)
    frames = []
    for site, district, since in SITES:
        local = np.round(city_temp + rng.normal(0, 0.4, len(hours)) + rng.uniform(-0.5, 0.5), 1)
        tie = _tie(hours, rng)
        south = district == "South"
        # South: denser blocks and older wiring, so load climbs faster in a short hot spell
        prm = dict(base=44.0 if south else 42.0, ac=1.2 if south else 1.15, ac_start=23.5 if south else 24.0,
                   ac_power=1.45, share=0.86 if south else 0.84, growth=0.02)
        neighbour, loading = _simulate(hours, local, tie, rng, prm)
        true_load = loading + rng.normal(0, 0.8, len(loading))
        # The overload label comes from the true loading; what each logger reads is another matter (Lab 9):
        # the southern loggers were installed with the wrong CT ratio setting and read about 10% low.
        seen = true_load * (0.90 if south else 1.0) + rng.normal(0, 2.5 if south else 0.0, len(loading))
        d = pd.DataFrame({"timestamp": hours, "temp_c": local, "neighbour_load_pct": neighbour,
                          "tie_closed": tie, "loading_pct": np.round(np.clip(seen, 0, None), 1)})
        d["temp_3h_c"] = d.temp_c.rolling(3, min_periods=1).mean()
        d["temp_24h_c"] = d.temp_c.rolling(24, min_periods=1).mean()
        d["hour"] = d.timestamp.dt.hour
        truth_s = pd.Series(true_load)
        fut = pd.concat([truth_s.shift(-k) for k in range(1, 7)], axis=1).max(axis=1)
        d["overload_in_6h"] = (fut > OVERLOAD_PCT).astype(int)
        d = d.iloc[:-6]
        d = d[d.timestamp >= pd.Timestamp(f"{since}-01-01")] if since >= 2021 else d
        d.insert(0, "logger_since", since)
        d.insert(0, "district", district)
        d.insert(0, "site", site)
        frames.append(d)
    return pd.concat(frames, ignore_index=True)


# ---------------------------------------------------------------------------
# Track B · thermal images (Lab 7): is there a hot spot on the equipment?
def _ironbow(x):
    """Map a temperature field scaled 0..1 to the false colours a thermal camera shows."""
    x = np.clip(x, 0, 1)
    r = np.clip(1.5 * x + 0.1, 0, 1)
    g = np.clip(2.2 * (x - 0.35), 0, 1)
    b = np.clip(0.6 - 1.8 * np.abs(x - 0.25), 0, 1) + np.clip(3 * (x - 0.85), 0, 1)
    return np.stack([r, g, np.clip(b, 0, 1)], axis=-1)


def make_thermal_images(n=1200, size=96, seed=SEED + 4):
    """Small false-colour thermal images (n, size, size, 3) of switchgear connections, taken on
    inspection rounds at three substations. Label 1 = hot spot (a connection far hotter than its
    neighbours), 0 = normal. A quarter of the normal images were taken in afternoon sun: the whole
    cabinet side is warm, but no connection stands out, the hard case. Each substation has its own
    cabinet layout and camera angle, which is why the data must be split by substation."""
    rng = np.random.default_rng(seed)
    yy, xx = np.mgrid[0:size, 0:size] / size
    images = np.zeros((n, size, size, 3), dtype=np.float32)
    labels = np.zeros(n, dtype=int)
    sites = np.array(["A", "B", "C"])[rng.integers(0, 3, n)]
    kinds = rng.choice(["hot", "normal", "sun"], size=n, p=[0.40, 0.36, 0.24])
    layout = {"A": (3, 0.30, 0.72), "B": (4, 0.25, 0.65), "C": (3, 0.35, 0.78)}   # bars, top, bottom
    for i in range(n):
        site, kind = sites[i], kinds[i]
        bars, top, bottom = layout[site]
        amb = rng.uniform(0.10, 0.30)
        rng.standard_normal((size, size))                                # an unused draw, kept so the images stay the same
        field = np.full((size, size), amb) + 0.06 * yy
        shift = rng.normal(0, 0.03)
        xs = np.linspace(0.25, 0.75, bars) + shift
        load = rng.uniform(0.08, 0.22)                                   # every bar warms with the load
        for bx in xs:                                                    # vertical bus bars
            bar = (np.abs(xx - bx) < 0.035) & (yy > top) & (yy < bottom)
            field = np.where(bar, amb + load + 0.03 * rng.normal(), field)
        for bx in xs:                                                    # bolted connections on each bar
            for cy in (top + 0.05, bottom - 0.05):
                spot = ((xx - bx) / 0.04) ** 2 + ((yy - cy) / 0.03) ** 2 < 1
                field = np.where(spot, field + load * 0.4, field)
        if kind == "sun":                                                # a warm wash from one side, no single hot joint
            side = rng.choice([-1, 1])
            wash = np.clip(0.5 + side * (xx - 0.5) * 1.6, 0, 1) * rng.uniform(0.25, 0.45)
            field = field + wash
        if kind == "hot":                                                # one connection far hotter than the rest
            bx = rng.choice(xs)
            cy = rng.choice([top + 0.05, bottom - 0.05])
            r = rng.uniform(0.025, 0.05)
            heat = np.exp(-(((xx - bx) ** 2 + (yy - cy) ** 2) / (2 * r ** 2)))
            field = field + rng.uniform(0.35, 0.6) * heat
            labels[i] = 1
        if rng.random() < 0.35:                                          # a reflection off the cabinet door
            y0 = rng.uniform(0.1, 0.9)
            field = np.where(np.abs(yy - y0) < 0.015, field + 0.12, field)
        field = field + rng.normal(0, 0.02, field.shape)
        images[i] = _ironbow(field)
    flip = rng.random(n) < 0.03                                          # a few labels are simply wrong
    labels[flip] = 1 - labels[flip]
    return (images * 255).astype(np.uint8), labels, sites, kinds


# ---------------------------------------------------------------------------
# The cleaning and feature steps students build in Lab 2, as one reference function.
# Later labs load the clean table with it, so every lab starts from the same known state.
FEATURES = ["temp_c", "temp_3h_c", "temp_6h_c", "temp_24h_c", "neighbour_load_pct", "tie_closed",
            "loading_pct", "loading_1h_ago", "loading_change_1h", "loading_max_24h", "hour", "month", "hot_season"]


def prepare_transformer(raw):
    """Lab 2 in one function: mark missing readings, fix the temperature logged in tenths and the CT
    change, never fill across long gaps, drop the leaking column, add time and history columns, and
    add the six-hour-ahead answers."""
    d = raw.copy()
    d.loc[d["loading_pct"].isin([-999.0, 0.0]), "loading_pct"] = np.nan    # codes for 'no reading'
    tenths = d["temp_c"] > 100                                              # logged in tenths of a degree
    d.loc[tenths, "temp_c"] = d.loc[tenths, "temp_c"] / 10
    d = d.drop(columns="overload_report_time").set_index("timestamp").asfreq("h")  # the 11-day gap becomes visible
    d.loc[d.index >= "2022-11-19", "loading_pct"] /= 1.04                  # the new CT reads 4% high
    d["loading_was_missing"] = d["loading_pct"].isna().astype(int)
    d["loading_pct"] = d["loading_pct"].interpolate(limit=3, limit_area="inside")   # fill short holes only
    d["temp_3h_c"] = d["temp_c"].rolling(3).mean()
    d["temp_6h_c"] = d["temp_c"].rolling(6).mean()
    d["temp_24h_c"] = d["temp_c"].rolling(24).mean()
    d["loading_1h_ago"] = d["loading_pct"].shift(1)
    d["loading_change_1h"] = d["loading_pct"] - d["loading_1h_ago"]
    d["loading_max_24h"] = d["loading_pct"].rolling(24, min_periods=12).max()   # a day with some holes still counts
    d["hour"] = d.index.hour
    d["month"] = d.index.month
    d["hot_season"] = d.index.month.isin(HOT_MONTHS).astype(int)
    d["loading_in_6h"] = d["loading_pct"].shift(-6)
    ahead = pd.concat([d["loading_pct"].shift(-k) for k in range(1, 7)], axis=1)
    d["overload_in_6h"] = (ahead.max(axis=1) > OVERLOAD_PCT).astype(float).where(ahead.notna().all(axis=1))
    return d

history, test_2025 = make_transformer()
hourly = prepare_transformer(history)
print(f'Clean hourly table (the Lab 2 pipeline): {len(hourly):,} hours, {hourly.index.min():%b %Y} to {hourly.index.max():%b %Y}.')
```

Below about 27 °C (a day-long mean) the loading barely moves: the air conditioners are mostly off.
Above about 30 °C it climbs by roughly 3 to 4 points for every degree.
A straight line in temperature cannot draw that bend, but it can if you give it columns that bend:

| Column | What it lets the line do |
|---|---|
| `temp24h_squared` | curve upwards as the heat grows |
| `temp24h_above_30` | stay nearly flat below 30 °C, then climb (a "kink") |
| `temp3h_tie_open` | let heat count more when the tie switch is open for maintenance and this transformer carries its whole load |
| `hour_1` … `hour_23` | give every hour of the day its own level (the evening peak, the quiet dawn): a single weight on `hour` can only draw a straight slope across the day, never a daily cycle |

The model is still linear in its weights, so everything else stays the same.
The cell adds the columns **one group at a time**, so you see what each one buys.

تحت نحو 27 درجة (متوسط يوم كامل) بالكاد تتحرك نسبة التحميل: أجهزة التكييف مطفأة في الغالب.
وفوق نحو 30 درجة ترتفع بنحو 3 إلى 4 نقاط مع كل درجة.
الخط المستقيم في الحرارة لا يستطيع رسم هذا الانحناء، لكنه يستطيع إن أعطيته أعمدة تنحني:

| العمود | ما يتيحه للخط |
|---|---|
| ⁦`temp24h_squared`⁩ | أن ينحني إلى الأعلى كلما زاد الحر |
| ⁦`temp24h_above_30`⁩ | أن يبقى شبه مستوٍ تحت 30 درجة، ثم يصعد («انكسار»، kink) |
| ⁦`temp3h_tie_open`⁩ | أن يجعل الحر يُحسب أكثر حين يكون مفتاح الربط مفتوحاً للصيانة ويحمل هذا المحوّل حمله كله |
| ⁦`hour_1`⁩ … ⁦`hour_23`⁩ | أن يعطي كل ساعة من اليوم مستواها الخاص (ذروة المساء، وهدوء الفجر): الوزن الواحد على ⁦`hour`⁩ لا يرسم إلا ميلاً مستقيماً عبر اليوم، ولا يرسم دورة يومية أبداً |

النموذج ما زال خطياً في أوزانه، لذلك يبقى كل شيء آخر كما هو.
والخلية تضيف الأعمدة **مجموعةً بعد مجموعة**، فترى ما تضيفه كل منها.

</details>

```python
HOURS = [f"hour_{k}" for k in range(1, 24)]
PHYSICS = ["temp24h_squared", "temp24h_above_30", "temp3h_tie_open"] + HOURS
def physics_columns(t):
    t = t.copy()
    t["temp24h_squared"]  = t["temp_24h_c"] ** 2
    t["temp24h_above_30"] = (t["temp_24h_c"] - 30).clip(lower=0)
    t["temp3h_tie_open"]  = t["temp_3h_c"] * (1 - t["tie_closed"])
    hours = pd.DataFrame({h: (t["hour"] == k).astype(int) for k, h in enumerate(HOURS, start=1)}, index=t.index)
    return pd.concat([t, hours], axis=1)
train2, valid2 = physics_columns(train), physics_columns(valid)

steps, cols = {}, list(COLS)
for new in [None, ["temp24h_squared"], ["temp24h_above_30"], ["temp3h_tie_open"], HOURS]:
    if new is not None:
        cols = cols + new
    m = make_pipeline(StandardScaler(), LinearRegression()).fit(train2[cols], train2["loading_in_6h"])
    f = m.predict(valid2[cols])
    name = "the 13 columns" if new is None else ("+ hour_1 ... hour_23" if new is HOURS else "+ " + new[0])
    steps[name] = {**scores(valid2["loading_in_6h"], f),
                   "heat MAE": round(mean_absolute_error(valid2["loading_in_6h"][heat], f[heat]), 1)}
pd.DataFrame(steps).T
```

Read it from top to bottom.
The first two temperature bends take a little off every error, the heat hours included, as the curve promised; the tie-switch column adds almost nothing on top.
The big step is the hour columns: once each hour has its own level, the average miss drops by about a fifth, and the heat-hour miss drops again.

For the straight line, the daily cycle is a larger blind spot than the heat curve.
Even so, the heat hours, the fast climbs of hot summer mornings, are still missed by more than twice the average. What is left has other causes, and Part 6 shows them.

More columns also mean more chances for twins to fight.
**Ridge** adds a penalty `α × (sum of squared weights)` to the error, which keeps weights small and calm.
**Lasso** uses `α × (sum of absolute weights)` and pushes the weakest weights to exactly zero. The strength `α` is chosen on **validation**, never on training.

```python
target_time = valid2.index.shift(6, freq="h")                 # the hour each forecast is for
in_ramadan = (target_time >= "2024-03-11") & (target_time < "2024-04-10")
evening = target_time.hour.isin([20, 21, 22, 23, 0, 1])
ramadan_evening = pd.Series(np.asarray(in_ramadan) & np.asarray(evening), index=valid2.index)
other_evening = pd.Series(~np.asarray(in_ramadan) & np.asarray(evening), index=valid2.index)
print("Ramadan-evening hours:", ramadan_evening.sum())

ramadan_resid = None   # ✏️ the average residual on the Ramadan-evening hours
check("ramadan_resid", ramadan_resid)
print(f"for comparison, the average residual on all other evenings: {resid[other_evening].mean():+.1f} points")

resid_reading = ""   # ✏️ one line: what do the worst hours in the table have in common?
```

```python
# 🟢 The Ramadan evenings one by one, and the average residual by hour of the day
per_evening = resid[ramadan_evening].groupby(target_time[np.asarray(ramadan_evening)].normalize()).mean()
print(f"Ramadan evenings with a positive average residual: {(per_evening > 0).sum()} of {len(per_evening)}")

by_hour = pd.DataFrame({"Ramadan days": resid[pd.Series(np.asarray(in_ramadan), index=valid2.index)].groupby(target_time[np.asarray(in_ramadan)].hour).mean(),
                        "other days": resid[pd.Series(~np.asarray(in_ramadan), index=valid2.index)].groupby(target_time[~np.asarray(in_ramadan)].hour).mean()})
by_hour.plot(figsize=(8, 3.5), marker="o")
plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("hour of the day the forecast is for"); plt.ylabel("average residual (points)")
plt.title("Average residual by hour: Ramadan days against other days")
plt.tight_layout(); plt.show()

# 🟢 The fix: a calendar column. Ramadan began on these dates in the training years too.
RAMADAN = [("2021-04-13", "2021-05-13"), ("2022-04-02", "2022-05-02"), ("2023-03-23", "2023-04-22"), ("2024-03-11", "2024-04-10")]
def ramadan_evening_column(t):
    when = t.index.shift(6, freq="h")
    r = np.zeros(len(t), dtype=bool)
    for start, end in RAMADAN:
        r |= np.asarray((when >= start) & (when < end))
    return (r & np.asarray(when.hour.isin([20, 21, 22, 23, 0, 1]))).astype(int)
train3, valid3 = train2.copy(), valid2.copy()
train3["ramadan_evening"], valid3["ramadan_evening"] = ramadan_evening_column(train2), ramadan_evening_column(valid2)
with_cal = make_pipeline(StandardScaler(), Ridge(alpha=alpha_use)).fit(train3[COLS2 + ["ramadan_evening"]], train3["loading_in_6h"])
resid3 = valid3["loading_in_6h"] - with_cal.predict(valid3[COLS2 + ["ramadan_evening"]])
print(f"average residual on Ramadan evenings: {resid[ramadan_evening].mean():+.1f} points without the column, "
      f"{resid3[ramadan_evening].mean():+.1f} with it")
```

Two things to see:
- **A run of misses every Ramadan evening.** The residual is positive on nearly every evening of Ramadan, by several points, while other evenings sit only about a point above zero.
  The real load after iftar is higher than the forecast, day after day. (That small positive shift on ordinary days is a clue too: the city adds customers every year, and the model learned on 2021 to 2023.)
  A run like that is not noise; it is a **missing calendar column**. The model has no way to know which days are Ramadan.
  Give it one yes/no column and the run almost disappears. The fix costs one line, because the calendar is known years ahead.
- **The worst single misses.** Read the tie-switch columns: in nearly all of the worst hours the tie switch is **open six hours later**, on a hot day.
  In about half of them it is closed now and **opens during those six hours**.
  Normally the closed tie lets the neighbouring feeder carry about a sixth of this transformer's load; when it opens for maintenance, this transformer suddenly carries its whole load.
  That opening is planned switching. The model sees the switch **now**; it cannot see a switching that has not happened yet.

**That second one the model cannot know from its inputs.** No weighting of past readings foresees a switching operation.
The fix is not a cleverer model. It is a new input the control room already has: the **planned switching and maintenance schedule** for the next hours.
That is an engineering conversation with the control room, and it improves the forecast in a way no tuning in this lab can.
<details><summary>💡 Why not just use a more powerful model?</summary>

A more flexible model (trees in Lab 5, networks in Lab 7) bends better around what it can see: the heat, the rising loading, the switch position now.
It still cannot see a switching that has not happened, or a Ramadan it was never told about.
When the residuals show misses driven by information that is not in the inputs, collect that information; do not tune.
</details>

## Key points
- **Three scores.** MAE is what the operator feels; RMSE punishes big misses; R² is the share of ups and downs explained, not "accuracy".
- **Baseline first.** Judge a model against its baseline, on all rows **and** on the rows that matter. A model can win clearly on the average and still miss badly where it counts.
- **Weights as sentences.** A weight is "how much the forecast moves when one column moves, the others fixed". A wrong sign means a bug, a leak, or twins fighting; read twins as a group.
- **Bends and penalties.** Columns that bend let a straight line follow physics (the heat curve, the daily cycle); Ridge keeps weights calm and Lasso suggests columns to drop. Choose the strength on validation.
- **Residuals.** A run of misses on a calendar pattern means a missing column you can add. Misses from information you do not have (a switching not yet made) mean you get the information, not tune.
- **AI summaries.** An AI summary can turn a weight into a cause and R² into accuracy. Check every sentence against the numbers, and remember who decides, so the report reaches the director accurately and clearly.
