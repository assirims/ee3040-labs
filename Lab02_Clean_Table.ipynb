````python
#@title 🟢 Load the course data (run this cell; no need to read it)
import numpy as np
import pandas as pd

SEED = 3040
OVERLOAD_PCT = 100.0
HOT_MONTHS = [5, 6, 7, 8, 9]


def _temperature(hours, rng):
    """Hourly air temperature in °C for a hot, dry city: a seasonal wave, a daily cycle that
    peaks mid-afternoon, day-to-day weather that persists, and a few summer heat waves."""
    doy = hours.dayofyear.values
    hour = hours.hour.values
    season = 27.0 + 11.0 * np.sin(2 * np.pi * (doy - 110) / 365.25)
    daily = 6.0 * np.cos(2 * np.pi * (hour - 15) / 24)
    days = (hours.normalize() - hours[0].normalize()).days.values
    n_days = days.max() + 1
    anom = np.zeros(n_days)
    for d in range(1, n_days):                                  # weather persists for days
        anom[d] = 0.8 * anom[d - 1] + rng.normal(0, 1.0)
    day_ts = pd.date_range(hours[0].normalize(), periods=n_days, freq="D")
    summer = np.isin(day_ts.month, [6, 7, 8])
    for s in np.flatnonzero(summer & (rng.random(n_days) < 2.2 / 92)):   # about two heat waves a summer
        anom[s:s + int(rng.integers(3, 8))] += rng.uniform(2.0, 4.0)
    t = season + daily + anom[days] + rng.normal(0, 0.5, len(hours))
    return np.round(t, 1)


def _tie(hours, rng):
    """Tie switch to the neighbouring feeder: closed (1) except during maintenance windows,
    about ten a year, 6 to 30 hours each, when this transformer carries the whole load."""
    n = len(hours)
    closed = np.ones(n, dtype=int)
    for s in np.flatnonzero(rng.random(n) < 10 / 8760):
        closed[s:s + int(rng.integers(6, 31))] = 0
    return closed


# residential load shape by hour, as a fraction of the daily peak (evening peak, quiet dawn)
_PROFILE = np.array([.62, .56, .52, .49, .47, .47, .50, .55, .60, .64, .67, .70,
                     .72, .74, .76, .78, .80, .83, .88, .93, .96, .94, .85, .73])

TRANSFORMER = dict(base=42.0, ac=1.15, ac_start=24.0, ac_power=1.45, share=0.84, growth=0.02)


def _ramadan(hours):
    """Approximate Ramadan months for 2021-2025 (the evening peak moves later and grows)."""
    starts = {2021: "2021-04-13", 2022: "2022-04-02", 2023: "2023-03-23", 2024: "2024-03-11", 2025: "2025-03-01"}
    r = np.zeros(len(hours), dtype=bool)
    for y, s in starts.items():
        s = pd.Timestamp(s)
        r |= (hours >= s) & (hours < s + np.timedelta64(30, "D"))
    return r


def _simulate(hours, temp, tie, rng, p=TRANSFORMER):
    """Loading in % of rating, and the neighbouring feeder's loading."""
    n = len(hours)
    t = pd.Series(temp)
    felt = (0.45 * t + 0.55 * t.rolling(24, min_periods=1).mean()).values   # buildings store heat
    ac = p["ac"] * np.clip(felt - p["ac_start"], 0, None) ** p["ac_power"]
    shape = _PROFILE[hours.hour.values]
    weekend = np.isin(hours.dayofweek.values, [4, 5])                        # Friday and Saturday
    ram = _ramadan(hours)
    evening = np.isin(hours.hour.values, [20, 21, 22, 23, 0, 1])
    base = p["base"] * shape * np.where(weekend, 1.06, 1.0) * np.where(ram & evening, 1.15, 1.0)
    noise = np.zeros(n)
    for i in range(1, n):                                                    # slow, correlated noise
        noise[i] = 0.9 * noise[i - 1] + rng.normal(0, 1.1)
    years = (hours.year.values - 2021) + hours.dayofyear.values / 365.0
    demand = (base + ac * (0.55 + 0.45 * shape) + noise) * (1 + p.get("growth", 0.02)) ** years   # new customers every year
    loading = demand * np.where(tie == 1, p["share"], 1.0)
    neighbour = np.roll(0.9 * demand, 1) + rng.normal(0, 2.0, n)              # the neighbouring feeder follows the same heat, an hour later
    return np.round(np.clip(neighbour, 0, None), 1), np.round(np.clip(loading, 0, None), 1)


def _overload_events(loading):
    above = loading > OVERLOAD_PCT
    return int(np.sum(above[1:] & ~above[:-1]) + (1 if above[0] else 0))


def _overload_report_column(df):
    """Filled only on days the transformer overloaded: the time the control room logged the overload."""
    over = df["loading_pct"] > OVERLOAD_PCT
    first = df.loc[over].groupby(df.loc[over, "timestamp"].dt.date)["timestamp"].min()
    report = pd.Series("", index=df.index, dtype=object)
    day = df["timestamp"].dt.date
    for d, ts in first.items():
        report[day == d] = (ts + np.timedelta64(40, "m")).strftime("%H:%M")
    return report


def _add_mess(df, rng, gap=True):
    df = df.copy()
    n = len(df)
    val = df["loading_pct"].values.copy()
    # 1. bursts of the logger's 'no reading' code
    for s in rng.choice(n - 12, size=int(round(n * 0.0328 / 6)), replace=False):
        val[s:s + int(rng.integers(3, 10))] = -999.0
    # 2. dropouts to exactly zero
    for s in rng.choice(n - 14, size=max(2, int(round(n * 0.0092 / 9))), replace=False):
        val[s:s + int(rng.integers(6, 13))] = 0.0
    df["loading_pct"] = val
    # 3. three temperatures logged in tenths of a degree (history only)
    if gap:
        hot = np.flatnonzero((df["temp_c"].values > 45) & (df["temp_c"].values < 49))
        rows = rng.choice(hot, size=3, replace=False)
        df.loc[rows, "temp_c"] = (df.loc[rows, "temp_c"] * 10).round(0)
    # 4. the logger replacement: an 11-day gap, then a CT that reads 4% high (history only)
    if gap:
        cut0, cut1 = pd.Timestamp("2022-11-08"), pd.Timestamp("2022-11-19")
        df = df[(df["timestamp"] < cut0) | (df["timestamp"] >= cut1)].copy()
    after = df["timestamp"] >= pd.Timestamp("2022-11-19")
    real = df["loading_pct"] > 0
    df.loc[after & real, "loading_pct"] = (df.loc[after & real, "loading_pct"] * 1.04).round(1)
    return df.reset_index(drop=True)


def make_transformer(seed=SEED):
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2021-01-01", "2025-12-31 23:00", freq="h")
    temp = _temperature(hours, rng)
    tie = _tie(hours, rng)
    neighbour, loading = _simulate(hours, temp, tie, rng)
    df = pd.DataFrame({"timestamp": hours, "temp_c": temp, "neighbour_load_pct": neighbour,
                       "tie_closed": tie, "loading_pct": loading})
    df["temp_3h_c"] = df["temp_c"].rolling(3, min_periods=1).mean().round(1)
    df["overload_report_time"] = _overload_report_column(df)
    df = df[["timestamp", "temp_c", "temp_3h_c", "neighbour_load_pct", "tie_closed", "loading_pct", "overload_report_time"]]
    hist = df[df["timestamp"] < "2025-01-01"].reset_index(drop=True)
    test = df[df["timestamp"] >= "2025-01-01"].reset_index(drop=True)
    return _add_mess(hist, rng, gap=True), _add_mess(test, rng, gap=False)


# ---------------------------------------------------------------------------
# Track D · the asset register (Lab 5). One row per transformer; answer = faults next summer.
def make_assets(seed=SEED + 1, n=900):
    """900 distribution transformers with their asset-register columns and the faults they had next summer."""
    rng = np.random.default_rng(seed)
    district = rng.choice(["North", "South", "East"], size=n, p=[0.45, 0.30, 0.25])
    age = rng.integers(1, 41, n)
    rating = rng.choice([500, 1000, 1500], size=n, p=[0.30, 0.50, 0.20])
    peak = np.round(np.clip(rng.normal(78, 14, n), 35, 130), 1)
    customers = np.round(rating * rng.uniform(0.18, 0.32, n)).astype(int)
    cable_age = rng.integers(1, 36, n)
    acetylene = np.round(rng.gamma(1.2, 1.6, n) + 6 * (rng.random(n) < 0.07), 1)   # a gas from arcing inside the oil
    fans = (rng.random(n) < 0.35).astype(int)                                   # forced cooling fitted
    # Faults come from thresholds and interactions (overloaded and old, arcing gas, old cables under
    # heavy loading, no fans on the busiest units): easy for trees, hard for a straight line.
    lam = np.exp(-0.2 + 1.1 * (1.0 * ((peak > 95) & (age > 20)) + 0.9 * (acetylene > 5)
                               + 0.7 * (age > 30) + 0.5 * ((cable_age > 25) & (peak > 85))
                               + 0.4 * ((fans == 0) & (peak > 100)))
                 + 0.15 * np.log1p(customers / 50) + 0.01 * age + 0.1 * (district == "South"))
    faults = rng.poisson(lam)
    return pd.DataFrame({
        "transformer_id": [f"T{i:03d}" for i in range(1, n + 1)],
        "district": district,
        "age_years": age,
        "rating_kva": rating,
        "peak_loading_pct": peak,
        "customers": customers,
        "cable_age_years": cable_age,
        "acetylene_ppm": acetylene,
        "cooling_fans": fans,
        "faults_last_3_summers": rng.poisson(1.5 * np.sqrt(lam) * rng.uniform(0.3, 1.7, n)),   # history helps, noisily
        "survey_batch": rng.integers(1, 60, n),
        "crew_hours_next_summer": np.round(np.clip(faults * 11.5 + rng.normal(0, 3, n), 0, None)),
        "faults_next_summer": faults,
    })


# ---------------------------------------------------------------------------
# Track C · forty transformer load loggers (Lab 6). Hourly loading for 2024, no labels.
LOGGER_ZONES = ["residential"] * 10 + ["commercial"] * 14 + ["government"] * 10 + ["farms"] * 6

_ZONE_SHAPE = {   # loading by hour as a fraction of the zone's daily peak
    "residential": _PROFILE,
    "commercial": np.array([.25, .20, .18, .17, .17, .18, .22, .35, .60, .80, .90, .95,
                            .95, .92, .90, .93, .97, 1.0, 1.0, .98, .92, .80, .55, .35]),
    "government": np.array([.20, .18, .17, .17, .17, .20, .40, .80, .95, 1.0, 1.0, 1.0,
                            .98, .95, .70, .40, .30, .27, .25, .24, .23, .22, .21, .20]),
    "farms": np.array([.70, .72, .74, .75, .75, .72, .60, .50, .48, .50, .55, .60,
                       .65, .66, .66, .64, .60, .58, .60, .64, .66, .68, .69, .70]),
}
_ZONE_PEAK = {"residential": 85.0, "commercial": 80.0, "government": 70.0, "farms": 60.0}


def make_loggers(seed=SEED + 2):
    """Hourly loading of 40 transformer loggers in 2024 (long format), and the maintenance log
    the operator 'forgot to mention'."""
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2024-01-01", "2024-12-31 23:00", freq="h")
    n = len(hours)
    temp = _temperature(hours, rng)
    felt = pd.Series(temp).rolling(24, min_periods=1).mean().values
    heat = np.clip(felt - 22.0, 0, None) / 16.0
    frames = []
    for k in range(40):
        lid, zone = f"L{k + 1:02d}", LOGGER_ZONES[k]
        shape = _ZONE_SHAPE[zone][hours.hour.values]
        level = _ZONE_PEAK[zone] * rng.uniform(0.9, 1.1)
        load = level * shape * (0.45 + 0.55 * heat) + rng.normal(0, 1.2, n)
        frames.append(pd.DataFrame({"timestamp": hours, "logger_id": lid, "zone": zone, "loading_pct": load}))
    df = pd.concat(frames, ignore_index=True)
    # a few loggers get a problem each: finding them is the point of Lab 6
    stuck = (df.logger_id == "L17") & (df.timestamp >= "2024-02-03") & (df.timestamp < "2024-02-29")
    df.loc[stuck, "loading_pct"] = rng.normal(0.0, 0.2, int(stuck.sum()))
    df.loc[(df.logger_id == "L31") & (df.timestamp >= "2024-11-12"), "loading_pct"] += 4.0
    days = np.sort(rng.choice(np.arange(5, 360), size=14, replace=False))
    log = []
    for d in days:
        start = pd.Timestamp("2024-01-01") + np.timedelta64(int(d) * 24 + int(rng.integers(8, 12)), "h")
        end = start + np.timedelta64(int(rng.integers(4, 9)), "h")
        w = (df.logger_id == "L08") & (df.timestamp >= start) & (df.timestamp < end)
        df.loc[w, "loading_pct"] = np.abs(rng.normal(0.0, 0.3, int(w.sum())))
        log.append({"logger_id": "L08", "start": start, "end": end, "work": "switchgear maintenance, transformer de-energized"})
    df["loading_pct"] = df["loading_pct"].clip(lower=0).round(1)
    return df, pd.DataFrame(log)


# ---------------------------------------------------------------------------
# Twelve transformers in two districts (Lab 9): who does the model fail for?
SITES = ([(f"N{i}", "North", 2016 + i % 3) for i in range(1, 9)]
         + [(f"S{i}", "South", 2022) for i in range(1, 5)])


def make_network(seed=SEED + 3):
    """Hourly rows for 12 transformers, ready to model. Columns: site, district, logger_since,
    timestamp, the model inputs, and overload_in_6h (1 if the true loading passes 100% within the
    next six hours)."""
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2021-01-01", "2025-12-31 23:00", freq="h")
    city_temp = _temperature(hours, rng)
    frames = []
    for site, district, since in SITES:
        local = np.round(city_temp + rng.normal(0, 0.4, len(hours)) + rng.uniform(-0.5, 0.5), 1)
        tie = _tie(hours, rng)
        south = district == "South"
        # South: denser blocks and older wiring, so load climbs faster in a short hot spell
        prm = dict(base=44.0 if south else 42.0, ac=1.2 if south else 1.15, ac_start=23.5 if south else 24.0,
                   ac_power=1.45, share=0.86 if south else 0.84, growth=0.02)
        neighbour, loading = _simulate(hours, local, tie, rng, prm)
        true_load = loading + rng.normal(0, 0.8, len(loading))
        # The overload label comes from the true loading; what each logger reads is another matter (Lab 9):
        # the southern loggers were installed with the wrong CT ratio setting and read about 10% low.
        seen = true_load * (0.90 if south else 1.0) + rng.normal(0, 2.5 if south else 0.0, len(loading))
        d = pd.DataFrame({"timestamp": hours, "temp_c": local, "neighbour_load_pct": neighbour,
                          "tie_closed": tie, "loading_pct": np.round(np.clip(seen, 0, None), 1)})
        d["temp_3h_c"] = d.temp_c.rolling(3, min_periods=1).mean()
        d["temp_24h_c"] = d.temp_c.rolling(24, min_periods=1).mean()
        d["hour"] = d.timestamp.dt.hour
        truth_s = pd.Series(true_load)
        fut = pd.concat([truth_s.shift(-k) for k in range(1, 7)], axis=1).max(axis=1)
        d["overload_in_6h"] = (fut > OVERLOAD_PCT).astype(int)
        d = d.iloc[:-6]
        d = d[d.timestamp >= pd.Timestamp(f"{since}-01-01")] if since >= 2021 else d
        d.insert(0, "logger_since", since)
        d.insert(0, "district", district)
        d.insert(0, "site", site)
        frames.append(d)
    return pd.concat(frames, ignore_index=True)


# ---------------------------------------------------------------------------
# Track B · thermal images (Lab 7): is there a hot spot on the equipment?
def _ironbow(x):
    """Map a temperature field scaled 0..1 to the false colours a thermal camera shows."""
    x = np.clip(x, 0, 1)
    r = np.clip(1.5 * x + 0.1, 0, 1)
    g = np.clip(2.2 * (x - 0.35), 0, 1)
    b = np.clip(0.6 - 1.8 * np.abs(x - 0.25), 0, 1) + np.clip(3 * (x - 0.85), 0, 1)
    return np.stack([r, g, np.clip(b, 0, 1)], axis=-1)


def make_thermal_images(n=1200, size=96, seed=SEED + 4):
    """Small false-colour thermal images (n, size, size, 3) of switchgear connections, taken on
    inspection rounds at three substations. Label 1 = hot spot (a connection far hotter than its
    neighbours), 0 = normal. A quarter of the normal images were taken in afternoon sun: the whole
    cabinet side is warm, but no connection stands out, the hard case. Each substation has its own
    cabinet layout and camera angle, which is why the data must be split by substation."""
    rng = np.random.default_rng(seed)
    yy, xx = np.mgrid[0:size, 0:size] / size
    images = np.zeros((n, size, size, 3), dtype=np.float32)
    labels = np.zeros(n, dtype=int)
    sites = np.array(["A", "B", "C"])[rng.integers(0, 3, n)]
    kinds = rng.choice(["hot", "normal", "sun"], size=n, p=[0.40, 0.36, 0.24])
    layout = {"A": (3, 0.30, 0.72), "B": (4, 0.25, 0.65), "C": (3, 0.35, 0.78)}   # bars, top, bottom
    for i in range(n):
        site, kind = sites[i], kinds[i]
        bars, top, bottom = layout[site]
        amb = rng.uniform(0.10, 0.30)
        rng.standard_normal((size, size))                                # an unused draw, kept so the images stay the same
        field = np.full((size, size), amb) + 0.06 * yy
        shift = rng.normal(0, 0.03)
        xs = np.linspace(0.25, 0.75, bars) + shift
        load = rng.uniform(0.08, 0.22)                                   # every bar warms with the load
        for bx in xs:                                                    # vertical bus bars
            bar = (np.abs(xx - bx) < 0.035) & (yy > top) & (yy < bottom)
            field = np.where(bar, amb + load + 0.03 * rng.normal(), field)
        for bx in xs:                                                    # bolted connections on each bar
            for cy in (top + 0.05, bottom - 0.05):
                spot = ((xx - bx) / 0.04) ** 2 + ((yy - cy) / 0.03) ** 2 < 1
                field = np.where(spot, field + load * 0.4, field)
        if kind == "sun":                                                # a warm wash from one side, no single hot joint
            side = rng.choice([-1, 1])
            wash = np.clip(0.5 + side * (xx - 0.5) * 1.6, 0, 1) * rng.uniform(0.25, 0.45)
            field = field + wash
        if kind == "hot":                                                # one connection far hotter than the rest
            bx = rng.choice(xs)
            cy = rng.choice([top + 0.05, bottom - 0.05])
            r = rng.uniform(0.025, 0.05)
            heat = np.exp(-(((xx - bx) ** 2 + (yy - cy) ** 2) / (2 * r ** 2)))
            field = field + rng.uniform(0.35, 0.6) * heat
            labels[i] = 1
        if rng.random() < 0.35:                                          # a reflection off the cabinet door
            y0 = rng.uniform(0.1, 0.9)
            field = np.where(np.abs(yy - y0) < 0.015, field + 0.12, field)
        field = field + rng.normal(0, 0.02, field.shape)
        images[i] = _ironbow(field)
    flip = rng.random(n) < 0.03                                          # a few labels are simply wrong
    labels[flip] = 1 - labels[flip]
    return (images * 255).astype(np.uint8), labels, sites, kinds


# ---------------------------------------------------------------------------
# The cleaning and feature steps students build in Lab 2, as one reference function.
# Later labs load the clean table with it, so every lab starts from the same known state.
FEATURES = ["temp_c", "temp_3h_c", "temp_6h_c", "temp_24h_c", "neighbour_load_pct", "tie_closed",
            "loading_pct", "loading_1h_ago", "loading_change_1h", "loading_max_24h", "hour", "month", "hot_season"]

history, test_2025 = make_transformer()
df = history.copy()
print(f'Course data loaded: {len(df):,} hourly rows from {df.timestamp.min():%d %b %Y} to {df.timestamp.max():%d %b %Y}.')
````

# Part 2 · Missing values Python understands

Python has one special value for "no reading": `NaN` (*not a number*, written `np.nan`).
Every library knows it. Averages skip it, plots leave a hole, and models refuse it loudly instead of learning from it silently.

Start with the three impossible temperatures. They are **not** missing: they need evidence, not a blank.
A reading of 481 °C is not weather.

The logger also records its own **3-hour mean temperature**, `temp_3h_c`, and that mean **includes the hour itself**.
If the hour really were 481 °C, the mean of three hours would be at least 160 °C.
Three times the mean, minus the two hours before, gives back the hour itself. The cell prints that value next to each bad reading.

<details><summary>🌐 الترجمة العربية</summary>
<div dir="rtl" align="right">

**الجزء 2 · القيم المفقودة كما يفهمها بايثون**

لبايثون قيمة خاصة واحدة تعني «لا توجد قراءة»: ⁦`NaN`⁩ (*not a number*، أي «ليس رقماً»، وتُكتب ⁦`np.nan`⁩).
كل مكتبة تعرفها. المتوسطات تتخطاها، والرسوم تترك مكانها فارغاً، والنماذج ترفضها برسالة واضحة بدل أن تتعلم منها بصمت.

ابدأ بدرجات الحرارة الثلاث المستحيلة. فهي **ليست** مفقودة: تحتاج إلى دليل، لا إلى خانة فارغة.
قراءة 481 °م ليست طقساً.

يحفظ المسجّل أيضاً **متوسطه الخاص للحرارة في ثلاث ساعات**، ⁦`temp_3h_c`⁩، وهذا المتوسط **يشمل الساعة نفسها**.
لو كانت الساعة فعلاً 481 °م، لكان متوسط الساعات الثلاث 160 °م على الأقل.
وثلاثة أضعاف المتوسط، ناقص الساعتين السابقتين، يعيد الساعة نفسها. تطبع الخلية هذه القيمة بجانب كل قراءة خاطئة.

````python
bad = df["temp_c"] > 100
if not bad.any():
    print("no temperature above 100 °C left: this cell has already run")
implied = 3 * df["temp_3h_c"] - df["temp_c"].shift(1) - df["temp_c"].shift(2)   # the hour, rebuilt from the logger's own mean
print(pd.DataFrame({"timestamp": df["timestamp"], "temp_c as logged": df["temp_c"], "temp_3h_c": df["temp_3h_c"],
                    "the hour, from the 3-h mean": implied.round(1)})[bad].to_string(index=False))
df.loc[bad, "temp_c"] = df.loc[bad, "temp_c"] / 10        # logged in tenths of a degree
print("hottest hour now:", df["temp_c"].max(), "°C")
````

Each rebuilt hour is within 0.1 °C of one tenth of the logged value; the rest is the logger's rounding.
The two fields agree only if the logger wrote the temperature in **tenths of a degree**. So the fix is to divide by 10, not to delete the hour.
A fix backed by evidence is the only kind worth making: it keeps the table accurate and the decision clear.

Now the loading. Lab 1 showed two ways this table writes "no reading": **−999**, the logger's code, and **exactly 0**.
A transformer that feeds dense residential blocks never carries zero load. It reads 0 only during an **outage**, and an outage is a fact you ask the operator about, not a guess.

### ✏️ Your turn 3 · Correcting the new CT
The logger and CT installed on 19 November 2022 read **4 % high**: every reading since then is **1.04 times** the true loading.

### 3.1 Short holes only
Some holes are an hour or two: the logger blinked, the load did not. Filling them by drawing a straight line between the neighbours (*interpolation*) is safe.
Some holes are days long. A line across eleven days **invents** a loading nobody measured, and a model trained on it learns a calm that never happened.

The rule: fill at most **3 hours** in a row, only **between two real readings**, and keep a flag column that says which values were missing.
A logger that goes quiet is information too, not noise.

In pandas, `interpolate(limit=3, limit_area="inside")` does this. Know exactly what `limit=3` means: it fills **at most 3 hours in a row**.
A hole of 1 to 3 hours is closed completely. A longer hole gets only its first 3 hours filled, and the rest stays empty.
Those few hours at the edge of a long hole are a small, known compromise; a whole invented day is not.

# Part 4 · Columns the model cannot invent

A model only sees the columns you give it. The loading in six hours depends on three things:
**how hot it has been over the last hours** (buildings store heat, so air conditioning keeps working after the peak), **what the loading is doing now**, and **the time of day and the season**.
None of those is a single column in the raw table; you build them.

| New column | How | Why it helps |
|---|---|---|
| `temp_3h_c`, `temp_6h_c`, `temp_24h_c` | rolling mean of the last 3, 6 and 24 hours | the AC load follows the heat stored in the buildings, not just this hour's air |
| `loading_1h_ago`, `loading_change_1h` | the loading one hour ago, and the difference | a loading that climbs every hour is the strongest warning there is |
| `loading_max_24h` | the highest loading in the last 24 hours | a transformer that was near its rating yesterday afternoon will be again today |
| `hour`, `weekday`, `month`, `hot_season` | from the timestamp | the evening peak, the weekend, and the hot months (May to September) |

`rolling(6).mean()` averages the current hour and the five before it. `shift(1)` moves a column down one row, so each hour sees the value from an hour earlier.
Every hour now has its own row, so "one row earlier" really is "one hour earlier".

The cell also rebuilds `temp_3h_c` from the corrected temperature, so all three temperature means are made the same way.
The logger's own column has done its job as evidence.

# Part 5 · A split by time

A model is judged on data it has **never seen**. For a forecast, "never seen" means **the future**. So the split follows the calendar:

| Set | Years | Use |
|---|---|---|
| training | 2021, 2022, 2023 | the model learns here |
| validation | 2024 | you compare models and settings here |
| test | 2025 | locked until Lab 4; opened once, at the end |

The 2025 season is not in `df` at all: it was set aside from the start in a separate table, `test_2025`, so nothing you do today can touch it.

A random split puts some hours of each heat wave in training and the neighbouring hours in validation.

# Part 6 · Scaling without leakage

Temperature is about 5 to 52 °C, the loading about 10 to 120 % of rating, and the hour is 0 to 23.
Many models care about size: a column with big numbers drowns the others. **Standardizing** puts every column on the same footing:

`scaled = (value − mean) / standard deviation`

The result means *"how many spreads above or below average"*.
With a mean of 40 % and a spread of 20 %, a loading of 55 % becomes (55 − 40) / 20 = 0.75: three quarters of a spread above the usual loading.

**The rule:** the mean and the spread are measured on the **training rows only**. Validation and test are scaled with those same numbers.
Measured on all the data, they would give the training rows information about 2024 and 2025. That is a quiet leak, and the validation score would flatter you.

# Part 7 · The pipeline

Doing each step by hand is how you learn it. Doing it by hand every time is how mistakes creep in: a scaler fitted on the wrong rows, a fill forgotten on the test set.
A **pipeline** chains the steps into one object with two verbs:
- `fit(train)` learns everything it needs (medians for filling, means and spreads for scaling) **from the training rows only**;
- `transform(any rows)` applies exactly those learned numbers to any rows: validation, test, or tomorrow's live readings.

This pipeline has two steps. `StandardScaler` does what you just did by hand, for every column.
`SimpleImputer(strategy="median")` fills a missing value with the training median of its column.

On `train` it has nothing to fill: you dropped every incomplete row in Your turn 5. It is there for the **live feed**.
In the field a logger can drop out at the very hour you need a forecast, and the pipeline must still hand the model a complete row. The last lines of the cell show it doing that.

The pipeline is also the object you save and deploy in Lab 10.
The control room then gets exactly the preparation the model was trained with, consistently and safely.

<details><summary>🌐 الترجمة العربية</summary>
<div dir="rtl" align="right">

**الجزء 7 · الـ pipeline**

أداء كل خطوة بيدك هو طريقة تعلّمها. وأداؤها بيدك في كل مرة هو طريقة تسلّل الأخطاء: مقياس (scaler) مُدرَّب على الصفوف الخطأ، أو ملء نُسي في مجموعة الاختبار.
الـ **pipeline** يسلسل الخطوات في كائن واحد له فعلان:
- ⁦`fit(train)`⁩ يتعلم كل ما يحتاجه (الوسيطات للملء، والمتوسطات والانحرافات لتوحيد المقياس) **من صفوف التدريب فقط**؛
- ⁦`transform(any rows)`⁩ يطبّق تلك الأرقام المتعلَّمة نفسها بالضبط على أي صفوف: التحقق، أو الاختبار، أو قراءات الغد الحية.

للـ pipeline هنا خطوتان. ⁦`StandardScaler`⁩ يفعل ما فعلته للتو بيدك، لكل عمود.
و⁦`SimpleImputer(strategy="median")`⁩ يملأ القيمة المفقودة بوسيط عمودها في التدريب.

على ⁦`train`⁩ لا يجد ما يملؤه: فقد حذفت كل صف ناقص في دورك 5. وجوده من أجل **القراءات الحية**.
في الميدان قد يصمت المسجّل في الساعة نفسها التي تحتاج فيها إلى تنبؤ، ويجب أن يسلّم الـ pipeline النموذجَ صفاً كاملاً مع ذلك. آخر أسطر الخلية تُريك ذلك.

والـ pipeline أيضاً هو الكائن الذي تحفظه وتنشره في اللاب 10.

````python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

prepare = make_pipeline(SimpleImputer(strategy="median"), StandardScaler())
prepare.fit(train[MODEL_COLUMNS])                     # learns from 2021-2023 only

X_train = prepare.transform(train[MODEL_COLUMNS])
X_valid = prepare.transform(valid[MODEL_COLUMNS])
i = MODEL_COLUMNS.index("loading_pct")
print("training columns now have mean", X_train.mean(axis=0).round(2)[:4], "...")
print("validation columns have mean  ", X_valid.mean(axis=0).round(2)[:4], "...")
print(f"loading_pct: mean {X_train[:, i].mean():.2f} in training, {X_valid[:, i].mean():.2f} in validation")

live = valid[MODEL_COLUMNS].iloc[[0]].copy()
live["loading_pct"] = np.nan                          # a live reading that never arrived
print(f"a live row with no loading: filled with the training median, {prepare[0].statistics_[i]:.1f} %; "
      f"any NaN left after the pipeline? {bool(np.isnan(prepare.transform(live)).any())}")
print(f"spread of loading_pct: pandas .std() {train['loading_pct'].std():.4f} %, StandardScaler {prepare[-1].scale_[i]:.4f} %")
````
