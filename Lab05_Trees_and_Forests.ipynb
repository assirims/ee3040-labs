# Part 1 · Information gain, by hand

A decision tree asks yes/no questions: *is the fan set working? is the loading above 80 %?*
It picks the question that makes the answers **purest**: groups that are all overloads or all without one.

**Entropy** measures purity. It is highest when a group is half and half (the most doubt), and 0 when a group is all one kind (no doubt).

The **information gain** of a question is the entropy before it minus the entropy after it. *After* is a **weighted average**: each group counts in proportion to its size.

gain = entropy before − ( share of windows on the left × entropy of the left group + share on the right × entropy of the right group )

The weights matter. A pure group of 1 window counts less than a mixed group of 7.

The table holds eight six-hour windows on summer afternoons. Each comes from a distribution transformer of the same design, with forced-cooling fans fitted.
In the four windows with `fans_working` = no, the unit's fan set is out of service. The transformer then has only natural-air cooling, and a lower rating, to carry the load:

| window | loading now (% of rating) | `fans_working` | overload in the next 6 h |
|---|---|---|---|
| 1 | 85 | no | yes |
| 2 | 90 | no | yes |
| 3 | 78 | no | yes |
| 4 | 72 | no | yes |
| 5 | 88 | yes | no |
| 6 | 82 | yes | no |
| 7 | 75 | yes | no |
| 8 | 69 | yes | no |

Two candidate first questions: **"loading above 80 %?"** and **"fans working?"**. The next cell provides an `entropy(yes, no)` function.

````python
import numpy as np
def entropy(yes, no):
    total = yes + no
    return -sum(p * np.log2(p) for p in (yes / total, no / total) if p > 0)

entropy_before = None   # ✏️ entropy of 4 overloads and 4 windows without one
gain_loading = None     # ✏️ entropy_before minus the weighted entropy of the two loading groups
gain_fans = None        # ✏️ entropy_before minus the weighted entropy of the two fans groups
check("entropy_before", entropy_before)
check("gain_loading", gain_loading)
check("gain_fans", gain_fans)
````

Loading above 80 % tells you **nothing** here (gain 0): each side is still half overloads. Whether the fans are working tells you **everything** (gain 1 bit).
So the tree asks about the fans first. The loading only matters *inside* each branch.

````python
#@title 🟢 Load the course data (run this cell; no need to read it)
import numpy as np
import pandas as pd

SEED = 3040
OVERLOAD_PCT = 100.0
HOT_MONTHS = [5, 6, 7, 8, 9]


def _temperature(hours, rng):
    """Hourly air temperature in °C for a hot, dry city: a seasonal wave, a daily cycle that
    peaks mid-afternoon, day-to-day weather that persists, and a few summer heat waves."""
    doy = hours.dayofyear.values
    hour = hours.hour.values
    season = 27.0 + 11.0 * np.sin(2 * np.pi * (doy - 110) / 365.25)
    daily = 6.0 * np.cos(2 * np.pi * (hour - 15) / 24)
    days = (hours.normalize() - hours[0].normalize()).days.values
    n_days = days.max() + 1
    anom = np.zeros(n_days)
    for d in range(1, n_days):                                  # weather persists for days
        anom[d] = 0.8 * anom[d - 1] + rng.normal(0, 1.0)
    day_ts = pd.date_range(hours[0].normalize(), periods=n_days, freq="D")
    summer = np.isin(day_ts.month, [6, 7, 8])
    for s in np.flatnonzero(summer & (rng.random(n_days) < 2.2 / 92)):   # about two heat waves a summer
        anom[s:s + int(rng.integers(3, 8))] += rng.uniform(2.0, 4.0)
    t = season + daily + anom[days] + rng.normal(0, 0.5, len(hours))
    return np.round(t, 1)


def _tie(hours, rng):
    """Tie switch to the neighbouring feeder: closed (1) except during maintenance windows,
    about ten a year, 6 to 30 hours each, when this transformer carries the whole load."""
    n = len(hours)
    closed = np.ones(n, dtype=int)
    for s in np.flatnonzero(rng.random(n) < 10 / 8760):
        closed[s:s + int(rng.integers(6, 31))] = 0
    return closed


# residential load shape by hour, as a fraction of the daily peak (evening peak, quiet dawn)
_PROFILE = np.array([.62, .56, .52, .49, .47, .47, .50, .55, .60, .64, .67, .70,
                     .72, .74, .76, .78, .80, .83, .88, .93, .96, .94, .85, .73])

TRANSFORMER = dict(base=42.0, ac=1.15, ac_start=24.0, ac_power=1.45, share=0.84, growth=0.02)


def _ramadan(hours):
    """Approximate Ramadan months for 2021-2025 (the evening peak moves later and grows)."""
    starts = {2021: "2021-04-13", 2022: "2022-04-02", 2023: "2023-03-23", 2024: "2024-03-11", 2025: "2025-03-01"}
    r = np.zeros(len(hours), dtype=bool)
    for y, s in starts.items():
        s = pd.Timestamp(s)
        r |= (hours >= s) & (hours < s + np.timedelta64(30, "D"))
    return r


def _simulate(hours, temp, tie, rng, p=TRANSFORMER):
    """Loading in % of rating, and the neighbouring feeder's loading."""
    n = len(hours)
    t = pd.Series(temp)
    felt = (0.45 * t + 0.55 * t.rolling(24, min_periods=1).mean()).values   # buildings store heat
    ac = p["ac"] * np.clip(felt - p["ac_start"], 0, None) ** p["ac_power"]
    shape = _PROFILE[hours.hour.values]
    weekend = np.isin(hours.dayofweek.values, [4, 5])                        # Friday and Saturday
    ram = _ramadan(hours)
    evening = np.isin(hours.hour.values, [20, 21, 22, 23, 0, 1])
    base = p["base"] * shape * np.where(weekend, 1.06, 1.0) * np.where(ram & evening, 1.15, 1.0)
    noise = np.zeros(n)
    for i in range(1, n):                                                    # slow, correlated noise
        noise[i] = 0.9 * noise[i - 1] + rng.normal(0, 1.1)
    years = (hours.year.values - 2021) + hours.dayofyear.values / 365.0
    demand = (base + ac * (0.55 + 0.45 * shape) + noise) * (1 + p.get("growth", 0.02)) ** years   # new customers every year
    loading = demand * np.where(tie == 1, p["share"], 1.0)
    neighbour = np.roll(0.9 * demand, 1) + rng.normal(0, 2.0, n)              # the neighbouring feeder follows the same heat, an hour later
    return np.round(np.clip(neighbour, 0, None), 1), np.round(np.clip(loading, 0, None), 1)


def _overload_events(loading):
    above = loading > OVERLOAD_PCT
    return int(np.sum(above[1:] & ~above[:-1]) + (1 if above[0] else 0))


def _overload_report_column(df):
    """Filled only on days the transformer overloaded: the time the control room logged the overload."""
    over = df["loading_pct"] > OVERLOAD_PCT
    first = df.loc[over].groupby(df.loc[over, "timestamp"].dt.date)["timestamp"].min()
    report = pd.Series("", index=df.index, dtype=object)
    day = df["timestamp"].dt.date
    for d, ts in first.items():
        report[day == d] = (ts + np.timedelta64(40, "m")).strftime("%H:%M")
    return report


def _add_mess(df, rng, gap=True):
    df = df.copy()
    n = len(df)
    val = df["loading_pct"].values.copy()
    # 1. bursts of the logger's 'no reading' code
    for s in rng.choice(n - 12, size=int(round(n * 0.0328 / 6)), replace=False):
        val[s:s + int(rng.integers(3, 10))] = -999.0
    # 2. dropouts to exactly zero
    for s in rng.choice(n - 14, size=max(2, int(round(n * 0.0092 / 9))), replace=False):
        val[s:s + int(rng.integers(6, 13))] = 0.0
    df["loading_pct"] = val
    # 3. three temperatures logged in tenths of a degree (history only)
    if gap:
        hot = np.flatnonzero((df["temp_c"].values > 45) & (df["temp_c"].values < 49))
        rows = rng.choice(hot, size=3, replace=False)
        df.loc[rows, "temp_c"] = (df.loc[rows, "temp_c"] * 10).round(0)
    # 4. the logger replacement: an 11-day gap, then a CT that reads 4% high (history only)
    if gap:
        cut0, cut1 = pd.Timestamp("2022-11-08"), pd.Timestamp("2022-11-19")
        df = df[(df["timestamp"] < cut0) | (df["timestamp"] >= cut1)].copy()
    after = df["timestamp"] >= pd.Timestamp("2022-11-19")
    real = df["loading_pct"] > 0
    df.loc[after & real, "loading_pct"] = (df.loc[after & real, "loading_pct"] * 1.04).round(1)
    return df.reset_index(drop=True)


def make_transformer(seed=SEED):
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2021-01-01", "2025-12-31 23:00", freq="h")
    temp = _temperature(hours, rng)
    tie = _tie(hours, rng)
    neighbour, loading = _simulate(hours, temp, tie, rng)
    df = pd.DataFrame({"timestamp": hours, "temp_c": temp, "neighbour_load_pct": neighbour,
                       "tie_closed": tie, "loading_pct": loading})
    df["temp_3h_c"] = df["temp_c"].rolling(3, min_periods=1).mean().round(1)
    df["overload_report_time"] = _overload_report_column(df)
    df = df[["timestamp", "temp_c", "temp_3h_c", "neighbour_load_pct", "tie_closed", "loading_pct", "overload_report_time"]]
    hist = df[df["timestamp"] < "2025-01-01"].reset_index(drop=True)
    test = df[df["timestamp"] >= "2025-01-01"].reset_index(drop=True)
    return _add_mess(hist, rng, gap=True), _add_mess(test, rng, gap=False)


# ---------------------------------------------------------------------------
# Track D · the asset register (Lab 5). One row per transformer; answer = faults next summer.
def make_assets(seed=SEED + 1, n=900):
    """900 distribution transformers with their asset-register columns and the faults they had next summer."""
    rng = np.random.default_rng(seed)
    district = rng.choice(["North", "South", "East"], size=n, p=[0.45, 0.30, 0.25])
    age = rng.integers(1, 41, n)
    rating = rng.choice([500, 1000, 1500], size=n, p=[0.30, 0.50, 0.20])
    peak = np.round(np.clip(rng.normal(78, 14, n), 35, 130), 1)
    customers = np.round(rating * rng.uniform(0.18, 0.32, n)).astype(int)
    cable_age = rng.integers(1, 36, n)
    acetylene = np.round(rng.gamma(1.2, 1.6, n) + 6 * (rng.random(n) < 0.07), 1)   # a gas from arcing inside the oil
    fans = (rng.random(n) < 0.35).astype(int)                                   # forced cooling fitted
    # Faults come from thresholds and interactions (overloaded and old, arcing gas, old cables under
    # heavy loading, no fans on the busiest units): easy for trees, hard for a straight line.
    lam = np.exp(-0.2 + 1.1 * (1.0 * ((peak > 95) & (age > 20)) + 0.9 * (acetylene > 5)
                               + 0.7 * (age > 30) + 0.5 * ((cable_age > 25) & (peak > 85))
                               + 0.4 * ((fans == 0) & (peak > 100)))
                 + 0.15 * np.log1p(customers / 50) + 0.01 * age + 0.1 * (district == "South"))
    faults = rng.poisson(lam)
    return pd.DataFrame({
        "transformer_id": [f"T{i:03d}" for i in range(1, n + 1)],
        "district": district,
        "age_years": age,
        "rating_kva": rating,
        "peak_loading_pct": peak,
        "customers": customers,
        "cable_age_years": cable_age,
        "acetylene_ppm": acetylene,
        "cooling_fans": fans,
        "faults_last_3_summers": rng.poisson(1.5 * np.sqrt(lam) * rng.uniform(0.3, 1.7, n)),   # history helps, noisily
        "survey_batch": rng.integers(1, 60, n),
        "crew_hours_next_summer": np.round(np.clip(faults * 11.5 + rng.normal(0, 3, n), 0, None)),
        "faults_next_summer": faults,
    })


# ---------------------------------------------------------------------------
# Track C · forty transformer load loggers (Lab 6). Hourly loading for 2024, no labels.
LOGGER_ZONES = ["residential"] * 10 + ["commercial"] * 14 + ["government"] * 10 + ["farms"] * 6

_ZONE_SHAPE = {   # loading by hour as a fraction of the zone's daily peak
    "residential": _PROFILE,
    "commercial": np.array([.25, .20, .18, .17, .17, .18, .22, .35, .60, .80, .90, .95,
                            .95, .92, .90, .93, .97, 1.0, 1.0, .98, .92, .80, .55, .35]),
    "government": np.array([.20, .18, .17, .17, .17, .20, .40, .80, .95, 1.0, 1.0, 1.0,
                            .98, .95, .70, .40, .30, .27, .25, .24, .23, .22, .21, .20]),
    "farms": np.array([.70, .72, .74, .75, .75, .72, .60, .50, .48, .50, .55, .60,
                       .65, .66, .66, .64, .60, .58, .60, .64, .66, .68, .69, .70]),
}
_ZONE_PEAK = {"residential": 85.0, "commercial": 80.0, "government": 70.0, "farms": 60.0}


def make_loggers(seed=SEED + 2):
    """Hourly loading of 40 transformer loggers in 2024 (long format), and the maintenance log
    the operator 'forgot to mention'."""
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2024-01-01", "2024-12-31 23:00", freq="h")
    n = len(hours)
    temp = _temperature(hours, rng)
    felt = pd.Series(temp).rolling(24, min_periods=1).mean().values
    heat = np.clip(felt - 22.0, 0, None) / 16.0
    frames = []
    for k in range(40):
        lid, zone = f"L{k + 1:02d}", LOGGER_ZONES[k]
        shape = _ZONE_SHAPE[zone][hours.hour.values]
        level = _ZONE_PEAK[zone] * rng.uniform(0.9, 1.1)
        load = level * shape * (0.45 + 0.55 * heat) + rng.normal(0, 1.2, n)
        frames.append(pd.DataFrame({"timestamp": hours, "logger_id": lid, "zone": zone, "loading_pct": load}))
    df = pd.concat(frames, ignore_index=True)
    # a few loggers get a problem each: finding them is the point of Lab 6
    stuck = (df.logger_id == "L17") & (df.timestamp >= "2024-02-03") & (df.timestamp < "2024-02-29")
    df.loc[stuck, "loading_pct"] = rng.normal(0.0, 0.2, int(stuck.sum()))
    df.loc[(df.logger_id == "L31") & (df.timestamp >= "2024-11-12"), "loading_pct"] += 4.0
    days = np.sort(rng.choice(np.arange(5, 360), size=14, replace=False))
    log = []
    for d in days:
        start = pd.Timestamp("2024-01-01") + np.timedelta64(int(d) * 24 + int(rng.integers(8, 12)), "h")
        end = start + np.timedelta64(int(rng.integers(4, 9)), "h")
        w = (df.logger_id == "L08") & (df.timestamp >= start) & (df.timestamp < end)
        df.loc[w, "loading_pct"] = np.abs(rng.normal(0.0, 0.3, int(w.sum())))
        log.append({"logger_id": "L08", "start": start, "end": end, "work": "switchgear maintenance, transformer de-energized"})
    df["loading_pct"] = df["loading_pct"].clip(lower=0).round(1)
    return df, pd.DataFrame(log)


# ---------------------------------------------------------------------------
# Twelve transformers in two districts (Lab 9): who does the model fail for?
SITES = ([(f"N{i}", "North", 2016 + i % 3) for i in range(1, 9)]
         + [(f"S{i}", "South", 2022) for i in range(1, 5)])


def make_network(seed=SEED + 3):
    """Hourly rows for 12 transformers, ready to model. Columns: site, district, logger_since,
    timestamp, the model inputs, and overload_in_6h (1 if the true loading passes 100% within the
    next six hours)."""
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2021-01-01", "2025-12-31 23:00", freq="h")
    city_temp = _temperature(hours, rng)
    frames = []
    for site, district, since in SITES:
        local = np.round(city_temp + rng.normal(0, 0.4, len(hours)) + rng.uniform(-0.5, 0.5), 1)
        tie = _tie(hours, rng)
        south = district == "South"
        # South: denser blocks and older wiring, so load climbs faster in a short hot spell
        prm = dict(base=44.0 if south else 42.0, ac=1.2 if south else 1.15, ac_start=23.5 if south else 24.0,
                   ac_power=1.45, share=0.86 if south else 0.84, growth=0.02)
        neighbour, loading = _simulate(hours, local, tie, rng, prm)
        true_load = loading + rng.normal(0, 0.8, len(loading))
        # The overload label comes from the true loading; what each logger reads is another matter (Lab 9):
        # the southern loggers were installed with the wrong CT ratio setting and read about 10% low.
        seen = true_load * (0.90 if south else 1.0) + rng.normal(0, 2.5 if south else 0.0, len(loading))
        d = pd.DataFrame({"timestamp": hours, "temp_c": local, "neighbour_load_pct": neighbour,
                          "tie_closed": tie, "loading_pct": np.round(np.clip(seen, 0, None), 1)})
        d["temp_3h_c"] = d.temp_c.rolling(3, min_periods=1).mean()
        d["temp_24h_c"] = d.temp_c.rolling(24, min_periods=1).mean()
        d["hour"] = d.timestamp.dt.hour
        truth_s = pd.Series(true_load)
        fut = pd.concat([truth_s.shift(-k) for k in range(1, 7)], axis=1).max(axis=1)
        d["overload_in_6h"] = (fut > OVERLOAD_PCT).astype(int)
        d = d.iloc[:-6]
        d = d[d.timestamp >= pd.Timestamp(f"{since}-01-01")] if since >= 2021 else d
        d.insert(0, "logger_since", since)
        d.insert(0, "district", district)
        d.insert(0, "site", site)
        frames.append(d)
    return pd.concat(frames, ignore_index=True)


# ---------------------------------------------------------------------------
# Track B · thermal images (Lab 7): is there a hot spot on the equipment?
def _ironbow(x):
    """Map a temperature field scaled 0..1 to the false colours a thermal camera shows."""
    x = np.clip(x, 0, 1)
    r = np.clip(1.5 * x + 0.1, 0, 1)
    g = np.clip(2.2 * (x - 0.35), 0, 1)
    b = np.clip(0.6 - 1.8 * np.abs(x - 0.25), 0, 1) + np.clip(3 * (x - 0.85), 0, 1)
    return np.stack([r, g, np.clip(b, 0, 1)], axis=-1)


def make_thermal_images(n=1200, size=96, seed=SEED + 4):
    """Small false-colour thermal images (n, size, size, 3) of switchgear connections, taken on
    inspection rounds at three substations. Label 1 = hot spot (a connection far hotter than its
    neighbours), 0 = normal. A quarter of the normal images were taken in afternoon sun: the whole
    cabinet side is warm, but no connection stands out, the hard case. Each substation has its own
    cabinet layout and camera angle, which is why the data must be split by substation."""
    rng = np.random.default_rng(seed)
    yy, xx = np.mgrid[0:size, 0:size] / size
    images = np.zeros((n, size, size, 3), dtype=np.float32)
    labels = np.zeros(n, dtype=int)
    sites = np.array(["A", "B", "C"])[rng.integers(0, 3, n)]
    kinds = rng.choice(["hot", "normal", "sun"], size=n, p=[0.40, 0.36, 0.24])
    layout = {"A": (3, 0.30, 0.72), "B": (4, 0.25, 0.65), "C": (3, 0.35, 0.78)}   # bars, top, bottom
    for i in range(n):
        site, kind = sites[i], kinds[i]
        bars, top, bottom = layout[site]
        amb = rng.uniform(0.10, 0.30)
        rng.standard_normal((size, size))                                # an unused draw, kept so the images stay the same
        field = np.full((size, size), amb) + 0.06 * yy
        shift = rng.normal(0, 0.03)
        xs = np.linspace(0.25, 0.75, bars) + shift
        load = rng.uniform(0.08, 0.22)                                   # every bar warms with the load
        for bx in xs:                                                    # vertical bus bars
            bar = (np.abs(xx - bx) < 0.035) & (yy > top) & (yy < bottom)
            field = np.where(bar, amb + load + 0.03 * rng.normal(), field)
        for bx in xs:                                                    # bolted connections on each bar
            for cy in (top + 0.05, bottom - 0.05):
                spot = ((xx - bx) / 0.04) ** 2 + ((yy - cy) / 0.03) ** 2 < 1
                field = np.where(spot, field + load * 0.4, field)
        if kind == "sun":                                                # a warm wash from one side, no single hot joint
            side = rng.choice([-1, 1])
            wash = np.clip(0.5 + side * (xx - 0.5) * 1.6, 0, 1) * rng.uniform(0.25, 0.45)
            field = field + wash
        if kind == "hot":                                                # one connection far hotter than the rest
            bx = rng.choice(xs)
            cy = rng.choice([top + 0.05, bottom - 0.05])
            r = rng.uniform(0.025, 0.05)
            heat = np.exp(-(((xx - bx) ** 2 + (yy - cy) ** 2) / (2 * r ** 2)))
            field = field + rng.uniform(0.35, 0.6) * heat
            labels[i] = 1
        if rng.random() < 0.35:                                          # a reflection off the cabinet door
            y0 = rng.uniform(0.1, 0.9)
            field = np.where(np.abs(yy - y0) < 0.015, field + 0.12, field)
        field = field + rng.normal(0, 0.02, field.shape)
        images[i] = _ironbow(field)
    flip = rng.random(n) < 0.03                                          # a few labels are simply wrong
    labels[flip] = 1 - labels[flip]
    return (images * 255).astype(np.uint8), labels, sites, kinds


# ---------------------------------------------------------------------------
# The cleaning and feature steps students build in Lab 2, as one reference function.
# Later labs load the clean table with it, so every lab starts from the same known state.
FEATURES = ["temp_c", "temp_3h_c", "temp_6h_c", "temp_24h_c", "neighbour_load_pct", "tie_closed",
            "loading_pct", "loading_1h_ago", "loading_change_1h", "loading_max_24h", "hour", "month", "hot_season"]

assets = make_assets()
print(f'Asset register loaded: {len(assets)} transformers, {assets.shape[1]} columns.')
````

Before any model, apply Lab 1 and Lab 2 to the columns themselves. One column must not be an input, and one is suspicious:
- `crew_hours_next_summer` is recorded **during the summer you are predicting**: a leak. The next cell drops it.
- `survey_batch` is a filing number with no physical meaning. It stays as an input **on purpose**, to show what the models do with it in Part 6.

No time here: each row is a transformer, not an hour. So a random split with a fixed seed is right. A quarter of the transformers go to validation.

**One column per district, all three of them.** The next cell turns `district` into yes/no columns, as in Lab 2, and keeps all three (`drop_first=False`).
For a tree that is the right choice: *is it East?* is then one question. With `drop_first=True`, East exists only as *not North and not South*, two questions deep.
The district's importance then splits over columns that each tell half the story.

A line is different. Its three district columns always add up to 1, exactly like its constant term, so their three weights cannot be told apart.
A line whose weights you want to read needs `drop_first=True`. The line in Part 2 shares this table.
Its predictions are not affected; only its district weights lose their meaning, and this lab does not read them.

# Part 5 · Boosting, neighbours, and comparison

**Gradient boosting** builds trees **one after another**. Each small tree learns the misses of all the trees before it, and joins with a small step: gradient descent where each step is a tree.
It often wins on tables, but it needs care. With too many trees it memorises too, so it stops early when a held-out slice stops improving.

**k-nearest neighbours** does no training at all. For a new transformer it finds the `k` most similar transformers in the training data and averages their faults.
Its strength is the explanation, not the score: *"here are the ten past transformers most like this one"*. It needs scaled columns, and it appears in the table below.

**Support vector machines** (SVM) draw the widest possible boundary between classes. They suit small, clean tables with many columns (the gas spectra of a transformer-oil test, for example).
On a table like this one they are rarely worth the tuning. Recognise the name and know when to reach for it; this lab does not train one.

<details><summary>🌐 الترجمة العربية</summary>
<div dir="rtl" align="right">

**الجزء 5 · التعزيز، والجيران، والمقارنة**

**التعزيز التدريجي** يبني الأشجار **واحدة بعد أخرى**. كل شجرة صغيرة تتعلم أخطاء كل الأشجار التي قبلها، وتُضاف بخطوة صغيرة: انحدار تدريجي (gradient descent) كل خطوة فيه شجرة.
وكثيراً ما يفوز على الجداول، لكنه يحتاج عناية. إن كثرت الأشجار حفظ هو أيضاً، لذلك يتوقف مبكراً حين يكفّ جزء محجوز من البيانات عن التحسن.

**أقرب k جيران** لا يتدرب إطلاقاً. لأي محوّل جديد يجد أكثر ⁦`k`⁩ محوّلات شبهاً به في بيانات التدريب، ويأخذ متوسط أعطالها.

**آلات متجهات الدعم** (SVM) ترسم أوسع حد ممكن بين الفئات. وتناسب الجداول الصغيرة النظيفة ذات الأعمدة الكثيرة (أطياف الغازات في اختبار زيت المحوّل مثلاً).
ونادراً ما يستحق ضبطها العناء على جدول كهذا. تعرّف على الاسم واعرف متى تلجأ إليها؛ وهذا اللاب لا يدرّب واحدة منها.

# Part 6 · Importance, leaks, and the budget

A forest reports how much each column helps. There are two ways to ask, and they disagree.
- **Built-in importance** adds up how much each column reduces the error inside the trees, **on the training data**. It costs nothing, but it favours columns with many possible cut points, whatever they mean.
- **Permutation importance** shuffles one column in the **validation** data and measures how much the score drops. If the model truly relies on a column, scrambling it hurts. It uses data the model has not seen, and it works for any model.

Permutation has one blind spot: **correlated columns share the credit**. When two columns carry the same information, scrambling one leaves the other to cover for it.
Each can then look unimportant, although together they matter.

**الجزء 6 · الأهمية، والتسرّب، والميزانية**

تخبرك الغابة كم يساعد كل عمود. وهناك طريقتان للسؤال، ولا تتفقان.
- **الأهمية المدمجة** (built-in importance) تجمع مقدار ما يخفّضه كل عمود من الخطأ داخل الأشجار، **على بيانات التدريب**. لا تكلّف شيئاً، لكنها تحابي الأعمدة ذات نقاط القطع الممكنة الكثيرة، أياً كان معناها.
- **أهمية التبديل** (permutation importance) تخلط عموداً واحداً في بيانات **التحقق** وتقيس كم تنخفض النتيجة. إن كان النموذج يعتمد على العمود حقاً، فخلطه يضر. وهي تستخدم بيانات لم يرها النموذج، وتصلح لأي نموذج.

لأهمية التبديل نقطة عمياء واحدة: **الأعمدة المترابطة تتقاسم الفضل**. حين يحمل عمودان المعلومة نفسها، فإن خلط أحدهما يترك الآخر يعوّضه.
فقد يبدو كل منهما غير مهم، مع أنهما معاً مهمان.

ويضيف الجدول التالي عموداً ثالثاً للمقارنة: ارتباط كل مدخل وحده بأعطال الصيف القادم.

</div>
</details>

````python
from sklearn.inspection import permutation_importance
builtin = pd.Series(forest.feature_importances_, index=X.columns)
perm = permutation_importance(forest, X_valid, y_valid, n_repeats=10, random_state=0)
importance = pd.DataFrame({"built-in (training)": builtin.round(3),
                           "permutation (validation)": pd.Series(perm.importances_mean, index=X.columns).round(3),
                           "correlation with faults": X.corrwith(y).round(2)})
importance.sort_values("permutation (validation)", ascending=False)
````

Look at two rows:
- `survey_batch`, a **filing number**, gets a built-in importance comparable to real columns like the customer count. Permutation shows its real worth: nothing. A column with many values offers many cut points, and the built-in measure rewards that.
- `faults_last_3_summers` is, on its own, one of the columns most correlated with next summer's faults, about as strongly as the age. Yet scrambling it costs the forest little. Its information is shared with the age, the peak loading and the acetylene, so the forest does without it.

That is the blind spot at work. A low permutation score means *the model manages without this column, given the others*, not *this column carries no information*.
To test a group of related columns, scramble them together, or drop them and retrain.

**Prefer permutation importance, on validation data, and read it with that blind spot in mind.** It tells you *what the model uses*, not *what causes faults*.

### The leak, revisited
The next cell adds `crew_hours_next_summer` back as an input. Run and watch the score and the top column.

<details><summary>🌐 الترجمة العربية</summary>
<div dir="rtl" align="right">

انظر إلى صفين:
- ⁦`survey_batch`⁩، وهو **رقم أرشفة**، يحصل على أهمية مدمجة تقارب أعمدة حقيقية مثل عدد المشتركين. وأهمية التبديل تبيّن قيمته الحقيقية: لا شيء. العمود ذو القيم الكثيرة يقدّم نقاط قطع كثيرة، والمقياس المدمج يكافئ ذلك.
- ⁦`faults_last_3_summers`⁩ هو، وحده، من أكثر الأعمدة ارتباطاً بأعطال الصيف القادم، بقوة تقارب العمر. ومع ذلك فخلطه لا يكلّف الغابة إلا قليلاً. معلوماته مشتركة مع العمر وذروة التحميل والأسيتيلين، فتستغني الغابة عنه.

هذه هي النقطة العمياء أمامك. انخفاض أهمية التبديل يعني *أن النموذج يستغني عن هذا العمود ما دامت الأعمدة الأخرى موجودة*، لا *أن العمود لا يحمل معلومة*.
ولاختبار مجموعة من الأعمدة المترابطة، اخلطها معاً، أو احذفها وأعد التدريب.

**فضّل أهمية التبديل، على بيانات التحقق، واقرأها وأنت تتذكر نقطتها العمياء.** فهي تخبرك *بما يستخدمه النموذج*، لا *بما يسبب الأعطال*.

**التسرّب، من جديد**

الخلية التالية تعيد إضافة ⁦`crew_hours_next_summer`⁩ مدخلاً. شغّل وراقب النتيجة والعمود الأول.

</div>
</details>

````python
X_leak = X.assign(crew_hours_next_summer=assets["crew_hours_next_summer"])
X_leak_train, X_leak_valid, y_leak_train, y_leak_valid = train_test_split(X_leak, y, test_size=0.25, random_state=0)
leaky = RandomForestRegressor(n_estimators=400, min_samples_leaf=3, random_state=0, n_jobs=-1).fit(X_leak_train, y_leak_train)
print("validation R² with the leaking column:", round(r2_score(y_leak_valid, leaky.predict(X_leak_valid)), 3))
print("most important column:", pd.Series(leaky.feature_importances_, index=X_leak.columns).idxmax())
````

Almost perfect, and useless. When the maintenance budget is set in spring, nobody knows how many crew hours next summer will take.
**A near-perfect score is a question, not a result.** The first question is always: which column is telling the model the answer?

### The maintenance budget: 200 of 900
The director does not need an R². The budget covers **200 transformers**, and the question is *which ones*.
So score the models on that decision: rank the transformers by forecast, fund the top 200, and count how many of next summer's faults they hold.

To rank all 900 fairly, every transformer needs a forecast from a model that did **not** see it. `cross_val_predict` does that.
It cuts the transformers into five folds and predicts each fold with a model trained on the other four.

<details><summary>🌐 الترجمة العربية</summary>
<div dir="rtl" align="right">

شبه مثالية، وبلا فائدة. حين تُحدَّد ميزانية الصيانة في الربيع، لا أحد يعرف كم ساعة عمل سيحتاجها الصيف القادم.
**النتيجة شبه المثالية سؤال، لا نتيجة.** والسؤال الأول دائماً: أي عمود يخبر النموذج بالإجابة؟

**ميزانية الصيانة: 200 من 900**

المدير لا يحتاج إلى R². الميزانية تغطي **200 محوّل**، والسؤال هو *أيها*.
فقيّم النماذج على هذا القرار: رتّب المحوّلات حسب التنبؤ، وموّل أعلى 200، واحسب كم تضم من أعطال الصيف القادم.

ولكي ترتّب المحوّلات التسعمئة كلها بإنصاف، يحتاج كل محوّل إلى تنبؤ من نموذج **لم** يره. وهذا ما تفعله ⁦`cross_val_predict`⁩.
فهي تقسم المحوّلات إلى خمس طيّات (folds)، وتتنبأ بكل طيّة بنموذج مدرَّب على الأربع الأخرى.

</div>
</details>

````python
from sklearn.model_selection import cross_val_predict, KFold
folds = KFold(n_splits=5, shuffle=True, random_state=0)
forecast = pd.DataFrame({
    "random forest": cross_val_predict(RandomForestRegressor(n_estimators=400, min_samples_leaf=3, random_state=0, n_jobs=-1), X, y, cv=folds),
    "straight line": cross_val_predict(make_pipeline(StandardScaler(), LinearRegression()), X, y, cv=folds),
    "last 3 summers (no model)": assets["faults_last_3_summers"],
    "perfect hindsight": y,
})
for rule, score in forecast.items():
    funded = score.sort_values(ascending=False, kind="stable").index[:200]
    print(f"{rule:27s} its top 200 transformers hold {y.loc[funded].sum() / y.sum():.0%} of next summer's faults")
print(f"{'200 transformers at random':27s} hold about {200 / len(y):.0%}")
````

Now the models speak the director's language. Both models beat the simple rule *"fund the transformers that failed most in the last three summers"*, and none reaches perfect hindsight.
Notice that the forest's clear lead in R² almost **disappears** here. At the top of the list, both models agree on the obvious transformers (old, heavily loaded, with arcing gas in the oil).

Is the small difference that remains worth the forest's extra cost? That is now a question in transformers and faults, and a decision maker can answer it with clarity and confidence.
**Always score a model on the decision it serves**, not only on R².

## What you learned
- A tree picks the question with the largest information gain (entropy before minus the size-weighted entropy after). It finds combinations a line cannot: whether the fans are working changes the meaning of the loading.
- For a tree, keep one yes/no column per category (`drop_first=False`); a line whose weights you read needs `drop_first=True`.
- One deep tree memorises (perfect on training, nearly worthless on validation). A shallow tree with a minimum leaf size explains itself in sentences that each rest on a real group.
- A random forest averages many different trees and is the safe first serious model for a table. Boosting can edge ahead but needs care; use it when the extra point changes a decision.
- k-nearest neighbours explains by example; SVMs suit small, clean, wide tables. Deep learning rarely wins on a table this size.
- Built-in importance is biased towards columns with many values. Permutation importance on validation is more reliable, but correlated columns share its credit. Importance is what the model uses, not what causes the result.
- A near-perfect score is a question: find the column that tells the model the answer.
