# Part 1 · The neuron: a line with a bend

A neuron does two things:
1. **Weighted sum:** `z = w1·x1 + w2·x2 + b`. This is exactly the straight line of Lab 3.
2. **Activation:** `a = f(z)`, a bend. Inside a network, the usual one is **ReLU**: `max(0, z)`. It keeps positive numbers and turns negatives into 0. At the output for yes/no, it is the **sigmoid** from Lab 4.

Why the bend? A line of a line is still a line. Without activations, a hundred layers collapse into one linear regression. With them, enough neurons can draw any curve.

**A tiny network.** Inputs `x = (1, 2)`. Two hidden neurons, one output:
- `h1 = ReLU(1·x1 + (−1)·x2 + 0.5)`
- `h2 = ReLU(0.5·x1 + 1·x2 − 1)`
- `z = 2·h1 + 1·h2 − 1`, and `output = sigmoid(z)`

The true label is **1** (hot spot). The loss is Lab 4's log loss: `−log(output)`.

# Part 2 · Convolution: a sliding filter

An image of 96 × 96 pixels with 3 colours is 27,648 numbers. In a dense layer, every neuron sees every number. It needs 27,648 weights **per neuron**, and it knows nothing about which pixels are neighbours.

A **convolution** uses a small **filter** that slides over the input. The filter responds to a local pattern: an edge, a bright blob, a stripe. The same few weights serve everywhere.

In one dimension, take a window of loading readings from a load logger (track C). Loading is the current as a % of the rating, so it rises and falls with the current.
Readings `[2, 5, 3, 1, 4]`, filter `[1, 0, −1]`. Put the filter on three readings, multiply and add; then slide one step.

This filter fires on a **falling edge** (big number, then small): a loading that drops. In an image, the filter is 3 × 3 and slides in two directions; the arithmetic is the same.
Training **learns the filter values**; you choose only how many filters and how big. Stacked layers build up step by step: edges → blobs and stripes → *"one joint much warmer than the joints beside it"*.

**A note on the name.** Slide, multiply and add, with the filter as it is, is strictly **cross-correlation** (hence `np.correlate` above). The textbook convolution first flips the filter end to end.
Keras's `Conv2D`, like every deep-learning library, does not flip it. Training learns the filter values anyway, so a flipped filter is simply learned flipped: only the name differs.

```python
#@title 🟢 Load the course data (run this cell; no need to read it)
import numpy as np
import pandas as pd

SEED = 3040
OVERLOAD_PCT = 100.0
HOT_MONTHS = [5, 6, 7, 8, 9]


def _temperature(hours, rng):
    """Hourly air temperature in °C for a hot, dry city: a seasonal wave, a daily cycle that
    peaks mid-afternoon, day-to-day weather that persists, and a few summer heat waves."""
    doy = hours.dayofyear.values
    hour = hours.hour.values
    season = 27.0 + 11.0 * np.sin(2 * np.pi * (doy - 110) / 365.25)
    daily = 6.0 * np.cos(2 * np.pi * (hour - 15) / 24)
    days = (hours.normalize() - hours[0].normalize()).days.values
    n_days = days.max() + 1
    anom = np.zeros(n_days)
    for d in range(1, n_days):                                  # weather persists for days
        anom[d] = 0.8 * anom[d - 1] + rng.normal(0, 1.0)
    day_ts = pd.date_range(hours[0].normalize(), periods=n_days, freq="D")
    summer = np.isin(day_ts.month, [6, 7, 8])
    for s in np.flatnonzero(summer & (rng.random(n_days) < 2.2 / 92)):   # about two heat waves a summer
        anom[s:s + int(rng.integers(3, 8))] += rng.uniform(2.0, 4.0)
    t = season + daily + anom[days] + rng.normal(0, 0.5, len(hours))
    return np.round(t, 1)


def _tie(hours, rng):
    """Tie switch to the neighbouring feeder: closed (1) except during maintenance windows,
    about ten a year, 6 to 30 hours each, when this transformer carries the whole load."""
    n = len(hours)
    closed = np.ones(n, dtype=int)
    for s in np.flatnonzero(rng.random(n) < 10 / 8760):
        closed[s:s + int(rng.integers(6, 31))] = 0
    return closed


# residential load shape by hour, as a fraction of the daily peak (evening peak, quiet dawn)
_PROFILE = np.array([.62, .56, .52, .49, .47, .47, .50, .55, .60, .64, .67, .70,
                     .72, .74, .76, .78, .80, .83, .88, .93, .96, .94, .85, .73])

TRANSFORMER = dict(base=42.0, ac=1.15, ac_start=24.0, ac_power=1.45, share=0.84, growth=0.02)


def _ramadan(hours):
    """Approximate Ramadan months for 2021-2025 (the evening peak moves later and grows)."""
    starts = {2021: "2021-04-13", 2022: "2022-04-02", 2023: "2023-03-23", 2024: "2024-03-11", 2025: "2025-03-01"}
    r = np.zeros(len(hours), dtype=bool)
    for y, s in starts.items():
        s = pd.Timestamp(s)
        r |= (hours >= s) & (hours < s + np.timedelta64(30, "D"))
    return r


def _simulate(hours, temp, tie, rng, p=TRANSFORMER):
    """Loading in % of rating, and the neighbouring feeder's loading."""
    n = len(hours)
    t = pd.Series(temp)
    felt = (0.45 * t + 0.55 * t.rolling(24, min_periods=1).mean()).values   # buildings store heat
    ac = p["ac"] * np.clip(felt - p["ac_start"], 0, None) ** p["ac_power"]
    shape = _PROFILE[hours.hour.values]
    weekend = np.isin(hours.dayofweek.values, [4, 5])                        # Friday and Saturday
    ram = _ramadan(hours)
    evening = np.isin(hours.hour.values, [20, 21, 22, 23, 0, 1])
    base = p["base"] * shape * np.where(weekend, 1.06, 1.0) * np.where(ram & evening, 1.15, 1.0)
    noise = np.zeros(n)
    for i in range(1, n):                                                    # slow, correlated noise
        noise[i] = 0.9 * noise[i - 1] + rng.normal(0, 1.1)
    years = (hours.year.values - 2021) + hours.dayofyear.values / 365.0
    demand = (base + ac * (0.55 + 0.45 * shape) + noise) * (1 + p.get("growth", 0.02)) ** years   # new customers every year
    loading = demand * np.where(tie == 1, p["share"], 1.0)
    neighbour = np.roll(0.9 * demand, 1) + rng.normal(0, 2.0, n)              # the neighbouring feeder follows the same heat, an hour later
    return np.round(np.clip(neighbour, 0, None), 1), np.round(np.clip(loading, 0, None), 1)


def _overload_events(loading):
    above = loading > OVERLOAD_PCT
    return int(np.sum(above[1:] & ~above[:-1]) + (1 if above[0] else 0))


def _overload_report_column(df):
    """Filled only on days the transformer overloaded: the time the control room logged the overload."""
    over = df["loading_pct"] > OVERLOAD_PCT
    first = df.loc[over].groupby(df.loc[over, "timestamp"].dt.date)["timestamp"].min()
    report = pd.Series("", index=df.index, dtype=object)
    day = df["timestamp"].dt.date
    for d, ts in first.items():
        report[day == d] = (ts + np.timedelta64(40, "m")).strftime("%H:%M")
    return report


def _add_mess(df, rng, gap=True):
    df = df.copy()
    n = len(df)
    val = df["loading_pct"].values.copy()
    # 1. bursts of the logger's 'no reading' code
    for s in rng.choice(n - 12, size=int(round(n * 0.0328 / 6)), replace=False):
        val[s:s + int(rng.integers(3, 10))] = -999.0
    # 2. dropouts to exactly zero
    for s in rng.choice(n - 14, size=max(2, int(round(n * 0.0092 / 9))), replace=False):
        val[s:s + int(rng.integers(6, 13))] = 0.0
    df["loading_pct"] = val
    # 3. three temperatures logged in tenths of a degree (history only)
    if gap:
        hot = np.flatnonzero((df["temp_c"].values > 45) & (df["temp_c"].values < 49))
        rows = rng.choice(hot, size=3, replace=False)
        df.loc[rows, "temp_c"] = (df.loc[rows, "temp_c"] * 10).round(0)
    # 4. the logger replacement: an 11-day gap, then a CT that reads 4% high (history only)
    if gap:
        cut0, cut1 = pd.Timestamp("2022-11-08"), pd.Timestamp("2022-11-19")
        df = df[(df["timestamp"] < cut0) | (df["timestamp"] >= cut1)].copy()
    after = df["timestamp"] >= pd.Timestamp("2022-11-19")
    real = df["loading_pct"] > 0
    df.loc[after & real, "loading_pct"] = (df.loc[after & real, "loading_pct"] * 1.04).round(1)
    return df.reset_index(drop=True)


def make_transformer(seed=SEED):
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2021-01-01", "2025-12-31 23:00", freq="h")
    temp = _temperature(hours, rng)
    tie = _tie(hours, rng)
    neighbour, loading = _simulate(hours, temp, tie, rng)
    df = pd.DataFrame({"timestamp": hours, "temp_c": temp, "neighbour_load_pct": neighbour,
                       "tie_closed": tie, "loading_pct": loading})
    df["temp_3h_c"] = df["temp_c"].rolling(3, min_periods=1).mean().round(1)
    df["overload_report_time"] = _overload_report_column(df)
    df = df[["timestamp", "temp_c", "temp_3h_c", "neighbour_load_pct", "tie_closed", "loading_pct", "overload_report_time"]]
    hist = df[df["timestamp"] < "2025-01-01"].reset_index(drop=True)
    test = df[df["timestamp"] >= "2025-01-01"].reset_index(drop=True)
    return _add_mess(hist, rng, gap=True), _add_mess(test, rng, gap=False)


# ---------------------------------------------------------------------------
# Track D · the asset register (Lab 5). One row per transformer; answer = faults next summer.
def make_assets(seed=SEED + 1, n=900):
    """900 distribution transformers with their asset-register columns and the faults they had next summer."""
    rng = np.random.default_rng(seed)
    district = rng.choice(["North", "South", "East"], size=n, p=[0.45, 0.30, 0.25])
    age = rng.integers(1, 41, n)
    rating = rng.choice([500, 1000, 1500], size=n, p=[0.30, 0.50, 0.20])
    peak = np.round(np.clip(rng.normal(78, 14, n), 35, 130), 1)
    customers = np.round(rating * rng.uniform(0.18, 0.32, n)).astype(int)
    cable_age = rng.integers(1, 36, n)
    acetylene = np.round(rng.gamma(1.2, 1.6, n) + 6 * (rng.random(n) < 0.07), 1)   # a gas from arcing inside the oil
    fans = (rng.random(n) < 0.35).astype(int)                                   # forced cooling fitted
    # Faults come from thresholds and interactions (overloaded and old, arcing gas, old cables under
    # heavy loading, no fans on the busiest units): easy for trees, hard for a straight line.
    lam = np.exp(-0.2 + 1.1 * (1.0 * ((peak > 95) & (age > 20)) + 0.9 * (acetylene > 5)
                               + 0.7 * (age > 30) + 0.5 * ((cable_age > 25) & (peak > 85))
                               + 0.4 * ((fans == 0) & (peak > 100)))
                 + 0.15 * np.log1p(customers / 50) + 0.01 * age + 0.1 * (district == "South"))
    faults = rng.poisson(lam)
    return pd.DataFrame({
        "transformer_id": [f"T{i:03d}" for i in range(1, n + 1)],
        "district": district,
        "age_years": age,
        "rating_kva": rating,
        "peak_loading_pct": peak,
        "customers": customers,
        "cable_age_years": cable_age,
        "acetylene_ppm": acetylene,
        "cooling_fans": fans,
        "faults_last_3_summers": rng.poisson(1.5 * np.sqrt(lam) * rng.uniform(0.3, 1.7, n)),   # history helps, noisily
        "survey_batch": rng.integers(1, 60, n),
        "crew_hours_next_summer": np.round(np.clip(faults * 11.5 + rng.normal(0, 3, n), 0, None)),
        "faults_next_summer": faults,
    })


# ---------------------------------------------------------------------------
# Track C · forty transformer load loggers (Lab 6). Hourly loading for 2024, no labels.
LOGGER_ZONES = ["residential"] * 10 + ["commercial"] * 14 + ["government"] * 10 + ["farms"] * 6

_ZONE_SHAPE = {   # loading by hour as a fraction of the zone's daily peak
    "residential": _PROFILE,
    "commercial": np.array([.25, .20, .18, .17, .17, .18, .22, .35, .60, .80, .90, .95,
                            .95, .92, .90, .93, .97, 1.0, 1.0, .98, .92, .80, .55, .35]),
    "government": np.array([.20, .18, .17, .17, .17, .20, .40, .80, .95, 1.0, 1.0, 1.0,
                            .98, .95, .70, .40, .30, .27, .25, .24, .23, .22, .21, .20]),
    "farms": np.array([.70, .72, .74, .75, .75, .72, .60, .50, .48, .50, .55, .60,
                       .65, .66, .66, .64, .60, .58, .60, .64, .66, .68, .69, .70]),
}
_ZONE_PEAK = {"residential": 85.0, "commercial": 80.0, "government": 70.0, "farms": 60.0}


def make_loggers(seed=SEED + 2):
    """Hourly loading of 40 transformer loggers in 2024 (long format), and the maintenance log
    the operator 'forgot to mention'."""
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2024-01-01", "2024-12-31 23:00", freq="h")
    n = len(hours)
    temp = _temperature(hours, rng)
    felt = pd.Series(temp).rolling(24, min_periods=1).mean().values
    heat = np.clip(felt - 22.0, 0, None) / 16.0
    frames = []
    for k in range(40):
        lid, zone = f"L{k + 1:02d}", LOGGER_ZONES[k]
        shape = _ZONE_SHAPE[zone][hours.hour.values]
        level = _ZONE_PEAK[zone] * rng.uniform(0.9, 1.1)
        load = level * shape * (0.45 + 0.55 * heat) + rng.normal(0, 1.2, n)
        frames.append(pd.DataFrame({"timestamp": hours, "logger_id": lid, "zone": zone, "loading_pct": load}))
    df = pd.concat(frames, ignore_index=True)
    # a few loggers get a problem each: finding them is the point of Lab 6
    stuck = (df.logger_id == "L17") & (df.timestamp >= "2024-02-03") & (df.timestamp < "2024-02-29")
    df.loc[stuck, "loading_pct"] = rng.normal(0.0, 0.2, int(stuck.sum()))
    df.loc[(df.logger_id == "L31") & (df.timestamp >= "2024-11-12"), "loading_pct"] += 4.0
    days = np.sort(rng.choice(np.arange(5, 360), size=14, replace=False))
    log = []
    for d in days:
        start = pd.Timestamp("2024-01-01") + np.timedelta64(int(d) * 24 + int(rng.integers(8, 12)), "h")
        end = start + np.timedelta64(int(rng.integers(4, 9)), "h")
        w = (df.logger_id == "L08") & (df.timestamp >= start) & (df.timestamp < end)
        df.loc[w, "loading_pct"] = np.abs(rng.normal(0.0, 0.3, int(w.sum())))
        log.append({"logger_id": "L08", "start": start, "end": end, "work": "switchgear maintenance, transformer de-energized"})
    df["loading_pct"] = df["loading_pct"].clip(lower=0).round(1)
    return df, pd.DataFrame(log)


# ---------------------------------------------------------------------------
# Twelve transformers in two districts (Lab 9): who does the model fail for?
SITES = ([(f"N{i}", "North", 2016 + i % 3) for i in range(1, 9)]
         + [(f"S{i}", "South", 2022) for i in range(1, 5)])


def make_network(seed=SEED + 3):
    """Hourly rows for 12 transformers, ready to model. Columns: site, district, logger_since,
    timestamp, the model inputs, and overload_in_6h (1 if the true loading passes 100% within the
    next six hours)."""
    rng = np.random.default_rng(seed)
    hours = pd.date_range("2021-01-01", "2025-12-31 23:00", freq="h")
    city_temp = _temperature(hours, rng)
    frames = []
    for site, district, since in SITES:
        local = np.round(city_temp + rng.normal(0, 0.4, len(hours)) + rng.uniform(-0.5, 0.5), 1)
        tie = _tie(hours, rng)
        south = district == "South"
        # South: denser blocks and older wiring, so load climbs faster in a short hot spell
        prm = dict(base=44.0 if south else 42.0, ac=1.2 if south else 1.15, ac_start=23.5 if south else 24.0,
                   ac_power=1.45, share=0.86 if south else 0.84, growth=0.02)
        neighbour, loading = _simulate(hours, local, tie, rng, prm)
        true_load = loading + rng.normal(0, 0.8, len(loading))
        # The overload label comes from the true loading; what each logger reads is another matter (Lab 9):
        # the southern loggers were installed with the wrong CT ratio setting and read about 10% low.
        seen = true_load * (0.90 if south else 1.0) + rng.normal(0, 2.5 if south else 0.0, len(loading))
        d = pd.DataFrame({"timestamp": hours, "temp_c": local, "neighbour_load_pct": neighbour,
                          "tie_closed": tie, "loading_pct": np.round(np.clip(seen, 0, None), 1)})
        d["temp_3h_c"] = d.temp_c.rolling(3, min_periods=1).mean()
        d["temp_24h_c"] = d.temp_c.rolling(24, min_periods=1).mean()
        d["hour"] = d.timestamp.dt.hour
        truth_s = pd.Series(true_load)
        fut = pd.concat([truth_s.shift(-k) for k in range(1, 7)], axis=1).max(axis=1)
        d["overload_in_6h"] = (fut > OVERLOAD_PCT).astype(int)
        d = d.iloc[:-6]
        d = d[d.timestamp >= pd.Timestamp(f"{since}-01-01")] if since >= 2021 else d
        d.insert(0, "logger_since", since)
        d.insert(0, "district", district)
        d.insert(0, "site", site)
        frames.append(d)
    return pd.concat(frames, ignore_index=True)


# ---------------------------------------------------------------------------
# Track B · thermal images (Lab 7): is there a hot spot on the equipment?
def _ironbow(x):
    """Map a temperature field scaled 0..1 to the false colours a thermal camera shows."""
    x = np.clip(x, 0, 1)
    r = np.clip(1.5 * x + 0.1, 0, 1)
    g = np.clip(2.2 * (x - 0.35), 0, 1)
    b = np.clip(0.6 - 1.8 * np.abs(x - 0.25), 0, 1) + np.clip(3 * (x - 0.85), 0, 1)
    return np.stack([r, g, np.clip(b, 0, 1)], axis=-1)


def make_thermal_images(n=1200, size=96, seed=SEED + 4):
    """Small false-colour thermal images (n, size, size, 3) of switchgear connections, taken on
    inspection rounds at three substations. Label 1 = hot spot (a connection far hotter than its
    neighbours), 0 = normal. A quarter of the normal images were taken in afternoon sun: the whole
    cabinet side is warm, but no connection stands out, the hard case. Each substation has its own
    cabinet layout and camera angle, which is why the data must be split by substation."""
    rng = np.random.default_rng(seed)
    yy, xx = np.mgrid[0:size, 0:size] / size
    images = np.zeros((n, size, size, 3), dtype=np.float32)
    labels = np.zeros(n, dtype=int)
    sites = np.array(["A", "B", "C"])[rng.integers(0, 3, n)]
    kinds = rng.choice(["hot", "normal", "sun"], size=n, p=[0.40, 0.36, 0.24])
    layout = {"A": (3, 0.30, 0.72), "B": (4, 0.25, 0.65), "C": (3, 0.35, 0.78)}   # bars, top, bottom
    for i in range(n):
        site, kind = sites[i], kinds[i]
        bars, top, bottom = layout[site]
        amb = rng.uniform(0.10, 0.30)
        rng.standard_normal((size, size))                                # an unused draw, kept so the images stay the same
        field = np.full((size, size), amb) + 0.06 * yy
        shift = rng.normal(0, 0.03)
        xs = np.linspace(0.25, 0.75, bars) + shift
        load = rng.uniform(0.08, 0.22)                                   # every bar warms with the load
        for bx in xs:                                                    # vertical bus bars
            bar = (np.abs(xx - bx) < 0.035) & (yy > top) & (yy < bottom)
            field = np.where(bar, amb + load + 0.03 * rng.normal(), field)
        for bx in xs:                                                    # bolted connections on each bar
            for cy in (top + 0.05, bottom - 0.05):
                spot = ((xx - bx) / 0.04) ** 2 + ((yy - cy) / 0.03) ** 2 < 1
                field = np.where(spot, field + load * 0.4, field)
        if kind == "sun":                                                # a warm wash from one side, no single hot joint
            side = rng.choice([-1, 1])
            wash = np.clip(0.5 + side * (xx - 0.5) * 1.6, 0, 1) * rng.uniform(0.25, 0.45)
            field = field + wash
        if kind == "hot":                                                # one connection far hotter than the rest
            bx = rng.choice(xs)
            cy = rng.choice([top + 0.05, bottom - 0.05])
            r = rng.uniform(0.025, 0.05)
            heat = np.exp(-(((xx - bx) ** 2 + (yy - cy) ** 2) / (2 * r ** 2)))
            field = field + rng.uniform(0.35, 0.6) * heat
            labels[i] = 1
        if rng.random() < 0.35:                                          # a reflection off the cabinet door
            y0 = rng.uniform(0.1, 0.9)
            field = np.where(np.abs(yy - y0) < 0.015, field + 0.12, field)
        field = field + rng.normal(0, 0.02, field.shape)
        images[i] = _ironbow(field)
    flip = rng.random(n) < 0.03                                          # a few labels are simply wrong
    labels[flip] = 1 - labels[flip]
    return (images * 255).astype(np.uint8), labels, sites, kinds


# ---------------------------------------------------------------------------
# The cleaning and feature steps students build in Lab 2, as one reference function.
# Later labs load the clean table with it, so every lab starts from the same known state.
FEATURES = ["temp_c", "temp_3h_c", "temp_6h_c", "temp_24h_c", "neighbour_load_pct", "tie_closed",
            "loading_pct", "loading_1h_ago", "loading_change_1h", "loading_max_24h", "hour", "month", "hot_season"]

images, hot_spot, image_site, image_kind = make_thermal_images()
print(f'Thermal images loaded: {len(images)} images of {images.shape[1]}x{images.shape[2]} pixels.')
```

Look closely. A thermal camera shows temperature as false colour: dark purple is cool, then red, orange, yellow, and white is hottest. The vertical stripes are the bus bars; the round patches are their bolted connections.

A **hot spot** is one connection far hotter than the others. A **sun-heated cabinet** (an outdoor kiosk or switchgear cabinet in the afternoon sun) is warm all over one side, but no single joint stands out. It is labelled 0, and it is the hard case.

Some images have a bright line: a reflection off the cabinet door. And a few labels are simply wrong, as in any real tagging.

**Split by substation, not by image.** Each substation has its own cabinet layout (three or four bars), its own distances and camera angle.
If images of the same substation sit on both sides of the split, the model can learn *the substation* instead of *the hot spot*, and the score lies (the same trap as neighbouring hours in Lab 3).
So: train on substations **A and B**, and validate on substation **C**, a substation the model has never seen. The validation score then speaks for a new substation, clearly and honestly.

انظر جيداً. الكاميرا الحرارية تعرض الحرارة بـ**ألوان زائفة** (false colour): البنفسجي الداكن بارد، ثم الأحمر والبرتقالي والأصفر، والأبيض هو الأسخن. والأشرطة العمودية هي القضبان الموصِّلة؛ والبقع المستديرة وصلاتها المربوطة.

# Part 4 · The old toolbox first

Before any network, try what Lab 5 does: make **columns by hand** and train a random forest. Two sets of columns:
- **Colour columns**: the average red, green and blue of the image, how varied its brightness is, and how much redder than blue it is.
- **The thermographer's columns**: the **hottest pixel**, and how far it stands **above the median** of the image. In these false colours, brightness rises with temperature. This is the thermographer's rule on site: a joint much hotter than its surroundings is suspect.

**A caution from the field.** Shiny, bare metal (a polished bus bar) has a low **emissivity** and reads cooler than it really is. Thermographers therefore compare like with like (joint against joint, phase against phase).
They also set the camera's emissivity, or put a patch of high-emissivity tape on the metal.

Then a plain **dense network** that sees every pixel (⏱ under a minute, even on the CPU).

<details><summary>🌐 الترجمة العربية</summary>
<div dir="rtl" align="right">

**الجزء 4 · صندوق الأدوات القديم أولاً**

قبل أي شبكة، جرّب ما يفعله اللاب 5: اصنع **أعمدة بيدك** ودرّب غابة عشوائية. مجموعتان من الأعمدة:
- **أعمدة الألوان**: متوسط الأحمر والأخضر والأزرق في الصورة، ومدى تفاوت سطوعها، وكم هي أكثر احمراراً من زرقتها.
- **أعمدة الفاحص الحراري**: **أسخن بكسل**، وكم يرتفع **فوق الوسيط** (median) في الصورة. وفي هذه الألوان الزائفة يزداد السطوع مع الحرارة. وهذه قاعدة الفاحص الحراري في الموقع: الوصلة الأسخن بكثير مما حولها مشتبه بها.

**تنبيه من الميدان.** المعدن اللامع المكشوف (قضيب موصِّل مصقول) **ابتعاثيته** (emissivity) منخفضة، فتقرؤه الكاميرا أبرد مما هو فعلاً. لذلك يقارن الفاحصون الشيء بمثله (وصلة بوصلة، وطوراً بطور).
ويضبطون كذلك الابتعاثية في الكاميرا، أو يلصقون على المعدن قطعة من شريط عالي الابتعاثية.

ثم **شبكة كثيفة** بسيطة ترى كل بكسل (⏱ أقل من دقيقة، حتى على المعالج العادي).

</div>
</details>

```python
from sklearn.ensemble import RandomForestClassifier
def colour_columns(images):
    return np.column_stack([images[..., 0].mean((1, 2)), images[..., 1].mean((1, 2)), images[..., 2].mean((1, 2)),
                            images.mean(-1).std((1, 2)), (images[..., 0] - images[..., 2]).mean((1, 2))])
def hottest_columns(images):
    brightness = images.mean(-1).reshape(len(images), -1)                 # one brightness number per pixel
    return np.column_stack([brightness.max(1), brightness.max(1) - np.median(brightness, 1)])
def hand_columns(images):
    return np.column_stack([colour_columns(images), hottest_columns(images)])

for name, make in [("colour columns", colour_columns), ("thermographer's columns", hottest_columns), ("both", hand_columns)]:
    f = RandomForestClassifier(n_estimators=300, random_state=0, n_jobs=-1).fit(make(X_train), y_train)
    right = f.predict(make(X_valid)) == y_valid
    by_kind = pd.Series(right).groupby(kind_valid).mean().mul(100).round(1)
    print(f"forest on {name:24s} {100 * right.mean():5.1f}% on substation C   by kind: {by_kind.to_dict()}")
forest = f                                                             # the last one: both sets of columns
forest_acc = 100 * right.mean()
```

```python
import keras
from keras import layers
keras.utils.set_random_seed(3040)

dense = keras.Sequential([keras.Input((96, 96, 3)), layers.Flatten(),          # 27,648 numbers in a row
                          layers.Dense(64, activation="relu"), layers.Dense(1, activation="sigmoid")])
dense.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
dense.fit(X_train, y_train, validation_data=(X_valid, y_valid), epochs=30, batch_size=32, verbose=0,
          callbacks=[keras.callbacks.EarlyStopping(patience=8, restore_best_weights=True)])
dense_acc = 100 * ((dense.predict(X_valid, verbose=0)[:, 0] > 0.5) == y_valid).mean()
print(f"dense network on raw pixels: {dense_acc:.1f}% on substation C, with {dense.count_params():,} weights")
```

Compare these numbers with your baseline from Your turn 3. The result may surprise you.
- **Colour columns** alone beat the baseline clearly, but miss many hot spots. One small hot joint changes the average colour of the image less than the load and the sun do.
- **The thermographer's columns** alone see most hot joints, but the **sun-heated cabinets** cost them: a warm cabinet side also has bright pixels well above the median. In the printed shares, their "sun" score falls to about 89%, where the colour columns reach about 97%. Their worst kind is actually the hot spots they miss (about 87%): a hot joint only a little warmer than the rest does not stand out enough.
- **Both together** are strong: the averages see the sun, the hottest pixel sees the joint. On these images the forest gets about 96 in 100 right. Remember that number.
- **The dense network** fails. Over several random starts it lands between about 60% and 67%, never clearly above always answering "no". It has almost 1.8 million weights and sees the pixels as an unordered list. To it, a hot joint on the left bar and a hot joint on the right bar are two unrelated patterns, and 785 images are far too few to learn each one.

So an engineer who knows what a hot spot is writes the column in one line, and here that column works.
The question for the rest of the lab: can a network **learn such columns from the pixels by itself**, without being told? That matters when nobody can write the column: a cracked insulator, a missing bolt, an oil stain on a bushing.

أربع كتل من *الالتفاف ← ReLU ← التجميع الأقصى*، ثم رأس كثيف صغير يقرر. ويحتفظ **التجميع** (pooling) بأقوى استجابة في كل مربع 2 × 2 ويقسم الحجم على اثنين. **اثنا عشر سطراً تقريباً من Keras.**
ثلاث من الكتل الأربع مكتوبة لك. والكتلة 1، أول ما ينظر إلى البكسلات، عليك.

**✏️ دورك 4 · الكتلة 1 وعدد الأوزان**

استبدل تعليق الكتلة 1 بالتفاف من 16 مرشّحاً بحجم 3 × 3 (مع ⁦`padding="same"`⁩ وReLU) يتبعه تجميع أقصى، ثم احفظ العدد الكلي لأوزان الشبكة في ⁦`n_weights`⁩.
<details><summary>💡 تلميح</summary>

اتخذ سطر الكتلة 2 نموذجاً: النوع نفسه من الطبقات، وحجم المرشّح نفسه والحشو نفسه. واستخدم نصف عدد المرشّحات، لأن الكتلة الأولى لا تبحث إلا عن حواف وبقع بسيطة. وكل نموذج في Keras يستطيع أن يعدّ أوزانه وانحيازاته بنفسه؛ والمجموع يظهر أيضاً في آخر الملخص. والكتلة الناقصة أو المختلفة تعطي مجموعاً مختلفاً، ويخبرك التحقق بذلك.
</details>
<details><summary>🔑 ما زلت عالقاً؟ الكود</summary>

الكتلة 1: ⁦`layers.Conv2D(16, 3, padding="same", activation="relu"), layers.MaxPooling2D(),`⁩ ثم ⁦`n_weights = cnn.count_params()`⁩.
</details>

```python
keras.utils.set_random_seed(3040)
cnn = keras.Sequential([
    keras.Input((96, 96, 3)),
    # ✏️ block 1: 16 filters of 3x3, padding "same", ReLU, then max pooling: edges and blobs
    layers.Conv2D(32, 3, padding="same", activation="relu"), layers.MaxPooling2D(),   # block 2: bars and joints
    layers.Conv2D(64, 3, padding="same", activation="relu"), layers.MaxPooling2D(),   # block 3: "a joint warmer than its bar"
    layers.Conv2D(64, 3, padding="same", activation="relu"), layers.MaxPooling2D(),   # block 4: the whole cabinet, "one joint stands out"
    layers.Flatten(), layers.Dropout(0.3),                                            # dropout: switch off 30% at random while training
    layers.Dense(32, activation="relu"),
    layers.Dense(1, activation="sigmoid"),                                            # yes/no: sigmoid, as in Lab 4
])
cnn.summary()
n_conv = sum(isinstance(layer, layers.Conv2D) for layer in cnn.layers)
if n_conv < 4:
    print(f"⚠️ The network has {n_conv} convolution blocks, not four. Write block 1, then run this cell again: the text below assumes four.")

n_weights = None   # ✏️ how many weights does the network have in total?
check("n_weights", n_weights)
```

Look at the summary: the four convolution layers together hold about 60,000 weights. The filters are small and shared across the whole image. The dense network, which failed, has 1.8 million.

🟢 Train it. **Early stopping** watches the validation loss and stops when it has not improved for 8 epochs, then restores the best weights. ⏱ About 1 to 3 minutes on the CPU, well under a minute on a GPU.

🟢 درّبها. **الإيقاف المبكر** (early stopping) يراقب خسارة التحقق ويتوقف حين لا تتحسن لمدة 8 حقب، ثم يستعيد أفضل الأوزان. ⏱ من دقيقة إلى ثلاث دقائق على المعالج العادي، وأقل من دقيقة بكثير على GPU.

```python
cnn.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])   # Adam, log loss (Lab 4), report accuracy
history = cnn.fit(X_train, y_train, validation_data=(X_valid, y_valid), epochs=30, batch_size=32, verbose=0,
                  callbacks=[keras.callbacks.EarlyStopping(patience=8, restore_best_weights=True)])
h = pd.DataFrame(history.history)
h[["loss", "val_loss"]].plot(figsize=(6, 3.2), title="Learning curves"); plt.xlabel("epoch"); plt.show()

cnn_pred = (cnn.predict(X_valid, verbose=0)[:, 0] > 0.5).astype(int)
cnn_acc = 100 * (cnn_pred == y_valid).mean()
train_acc = 100 * ((cnn.predict(X_train, verbose=0)[:, 0] > 0.5) == y_train).mean()   # the restored best weights, on the images they learned from
print(f"CNN from scratch: {cnn_acc:.1f}% on substation C, {train_acc:.1f}% on its own training images (substations A and B)")
print(pd.Series(cnn_pred == y_valid).groupby(kind_valid).mean().mul(100).round(1).rename("% correct by kind").to_string())
```

The deeper blocks combine these maps into the parts that decide *hot spot or not*. These are columns **learned from the pixels**.

# Part 6 · Transfer learning

**MobileNetV2** was trained by Google on ImageNet: over a million everyday photos in 1,000 classes (`weights="imagenet"`). Its layers already detect edges, blobs and textures. The code uses it in four steps:
- `include_top=False` removes its top, the last layer that named the 1,000 everyday objects.
- `pooling="avg"` averages each of its 1,280 final feature maps into one number, so every image becomes **1,280 learned columns**.
- `base.trainable = False` freezes it: nothing inside it is trained, and the code uses only what it already knows.
- On those 1,280 columns, the code trains **one new sigmoid neuron**: a logistic regression, exactly Lab 4's model. scikit-learn fits it (`LogisticRegression`), because that is exact and takes a second. A Keras `Dense(1, activation="sigmoid")` layer on top of the frozen network is the same neuron, trained step by step.

A fair doubt: ImageNet holds photos of dogs, cars and cups, not false-colour thermal images. Does what it learned still help here? Measure it; do not assume it.

🟢 ⏱ The first run downloads the pretrained weights (about 9 MB). Computing the columns takes about half a minute on the CPU.

<div dir="rtl" align="right">

لم يخبر أحدٌ الشبكةَ ما القضيب الموصِّل أو ما الوصلة الساخنة. بعض الخرائط تضيء على القضبان، وبعضها على حواف القضبان، وبعضها على أسخن بقعة فقط. وقد تبقى خريطة مظلمة حين لا يكون نمطها في هذه الصورة.
والكتل الأعمق تجمع هذه الخرائط في الأجزاء التي تقرر: *نقطة ساخنة أم لا*. هذه أعمدة **متعلَّمة من البكسلات**.

**الجزء 6 · التعلم بالنقل**

دُرّبت **MobileNetV2** في Google على ImageNet: أكثر من مليون صورة يومية في 1,000 فئة (⁦`weights="imagenet"`⁩). وطبقاتها تكتشف بالفعل الحواف والبقع والأنسجة. ويستخدمها الكود في أربع خطوات:
- ⁦`include_top=False`⁩ يحذف قمتها، أي الطبقة الأخيرة التي كانت تسمّي الأشياء اليومية الألف.
- ⁦`pooling="avg"`⁩ يأخذ متوسط كل خريطة من خرائط خصائصها الأخيرة البالغة 1,280 في رقم واحد، فتصبح كل صورة **1,280 عموداً متعلَّماً**.
- ⁦`base.trainable = False`⁩ يجمّدها: لا يُدرَّب شيء بداخلها، ويستخدم الكود فقط ما تعرفه.
- وعلى هذه الأعمدة الـ1,280 يدرّب الكود **عصبون sigmoid واحداً جديداً**: انحداراً لوجستياً، وهو بالضبط نموذج اللاب 4. ويدرّبه scikit-learn (⁦`LogisticRegression`⁩)، لأن ذلك دقيق ويستغرق ثانية. وطبقة ⁦`Dense(1, activation="sigmoid")`⁩ في Keras فوق الشبكة المجمَّدة هي العصبون نفسه، مدرَّباً خطوة بعد خطوة.

شكّ مشروع: في ImageNet صور كلاب وسيارات وأكواب، لا صور حرارية بألوان زائفة. فهل ما تعلمته ما زال يفيد هنا؟ قِسه، ولا تفترضه.

🟢 ⏱ التشغيل الأول ينزّل الأوزان **المدرَّبة مسبقاً** (pretrained) (نحو 9 MB). وحساب الأعمدة يستغرق نحو نصف دقيقة على المعالج العادي.

</div>
</details>

```python
base = keras.applications.MobileNetV2(input_shape=(96, 96, 3), include_top=False, weights="imagenet", pooling="avg")
base.trainable = False                                                     # frozen: we only use what it already knows
features = base.predict(keras.applications.mobilenet_v2.preprocess_input(images.astype("float32")), batch_size=64, verbose=0)
print("1,280 learned columns per image:", features.shape)

from sklearn.linear_model import LogisticRegression
F_train, F_valid = features[train], features[valid]
new_head = LogisticRegression(C=0.1, max_iter=3000).fit(F_train, y_train)   # the one new neuron: 1,280 weights and a bias
tl_pred = new_head.predict(F_valid)
tl_acc = 100 * (tl_pred == y_valid).mean()
print(f"transfer learning: {tl_acc:.1f}% on substation C   (your CNN from scratch: {cnn_acc:.1f}%)")
print(pd.Series(tl_pred == y_valid).groupby(kind_valid).mean().mul(100).round(1).rename("% correct by kind").to_string())
```

It still helps. Edges and blobs look alike in any picture, so the columns learned on everyday photos also describe a bright joint on a bar.
It takes seconds, trains not a single convolution, and gives the same result every run: the only thing trained is one logistic regression, which has no random start.
It lands within a point or two of your own CNN, which is usually a little ahead: 785 tagged images are enough to train it.

The real test is **few labels**. Tagging images costs a thermographer's time. What if only **150** were tagged?

Small data has a second tool: **augmentation**. Change a training image without changing its label: the network sees a new version every epoch, and learns the hot spot, not the cabinet.
For thermal images, the choice of changes is an engineering decision: **flips, a small rotation or shift, and at most a slight brightness jitter. Never change the colour palette or the contrast: in a thermal image, the colour is the temperature.**
Recolour a hot joint and you change the measurement, not the view.

The next cell shows what the network sees. **Training images only**: never augment validation or test, or you grade the model on images that do not exist.

<details><summary>🌐 الترجمة العربية</summary>
<div dir="rtl" align="right">

ما زال يفيد. فالحواف والبقع متشابهة في أي صورة، ولذلك فالأعمدة المتعلَّمة من الصور اليومية تصف أيضاً وصلة لامعة على قضيب.
يستغرق ذلك ثوانيَ، دون تدريب التفاف واحد، وبالنتيجة نفسها في كل تشغيل: فالشيء الوحيد المدرَّب انحدار لوجستي واحد، ولا بداية عشوائية له.
وتقع النتيجة على بُعد نقطة أو نقطتين من شبكتك الالتفافية الخاصة، التي تتقدم عليها قليلاً في الغالب: 785 صورة موسومة تكفي لتدريبها.

والاختبار الحقيقي هو **قلة التسميات**. وسم الصور يكلّف وقت فاحص حراري. فماذا لو لم يُوسَم إلا **150**؟

وللبيانات القليلة أداة ثانية: **التعزيز بالتحويرات** (augmentation). غيّر صورة التدريب دون أن تغيّر تسميتها: فترى الشبكة نسخة جديدة في كل حقبة، وتتعلم النقطة الساخنة، لا الخزانة.
وفي الصور الحرارية، اختيار التحويرات قرار هندسي: **القلب، وتدوير أو إزاحة صغيرة، وعلى الأكثر اهتزاز طفيف في السطوع. ولا تغيّر أبداً لوحة الألوان أو التباين: فاللون في الصورة الحرارية هو الحرارة.**
إن أعدت تلوين وصلة ساخنة فقد غيّرت القياس، لا زاوية النظر.

والخلية التالية تُظهر ما تراه الشبكة. **صور التدريب فقط**: لا تطبّق التعزيز بالتحويرات أبداً على التحقق أو الاختبار، وإلا فأنت تقيّم النموذج على صور غير موجودة.

```python
augment = keras.Sequential([layers.RandomFlip("horizontal"), layers.RandomRotation(0.03),       # the same cabinet, seen a little differently
                            layers.RandomTranslation(0.05, 0.05),
                            layers.RandomBrightness(0.05, value_range=(0, 1))])                 # a slight jitter only: colour = temperature
i = np.where((image_kind == "hot") & (hot_spot == 1))[0][2]
fig, axes = plt.subplots(1, 6, figsize=(12, 2.4))
axes[0].imshow(X[i]); axes[0].set_title("original", fontsize=9)
for ax in axes[1:]:
    ax.imshow(np.clip(augment(X[i:i + 1], training=True)[0], 0, 1)); ax.set_title("augmented", fontsize=9)
for ax in axes: ax.axis("off")
plt.show()
```

Keras applies augmentation only while training, so the substation C images are graded as they are. The augmented network sees a new version of each image every epoch, so it gets more epochs. ⏱ About 1 to 3 minutes on the CPU.

```python
few = np.random.default_rng(0).choice(np.where(train)[0], 150, replace=False)   # 150 tagged images from substations A and B
forest_few = 100 * (RandomForestClassifier(n_estimators=300, random_state=0, n_jobs=-1)
                    .fit(hand_columns(X[few]), hot_spot[few]).predict(hand_columns(X_valid)) == y_valid).mean()
tl_few = 100 * (LogisticRegression(C=0.1, max_iter=3000).fit(features[few], hot_spot[few]).predict(F_valid) == y_valid).mean()

keras.utils.set_random_seed(3040)
cnn_few = keras.models.clone_model(cnn)                  # the same shape as your CNN, with fresh random weights
cnn_few.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
cnn_few.fit(X[few], hot_spot[few], validation_data=(X_valid, y_valid), epochs=30, batch_size=32, verbose=0,
            callbacks=[keras.callbacks.EarlyStopping(patience=8, restore_best_weights=True)])
cnn_few_acc = 100 * ((cnn_few.predict(X_valid, verbose=0)[:, 0] > 0.5) == y_valid).mean()

keras.utils.set_random_seed(3040)
cnn_aug = keras.Sequential([keras.Input((96, 96, 3)), augment, keras.models.clone_model(cnn)])   # augmentation first, active in training only
cnn_aug.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
cnn_aug.fit(X[few], hot_spot[few], validation_data=(X_valid, y_valid), epochs=60, batch_size=32, verbose=0,
            callbacks=[keras.callbacks.EarlyStopping(patience=15, restore_best_weights=True)])
cnn_aug_acc = 100 * ((cnn_aug.predict(X_valid, verbose=0)[:, 0] > 0.5) == y_valid).mean()

results = pd.DataFrame({"substation C accuracy (%)": [acc_always_no, forest_acc, dense_acc, cnn_acc, tl_acc,
                                                       forest_few, cnn_few_acc, cnn_aug_acc, tl_few]},
    index=["always 'no hot spot'", "forest on hand-made columns", "dense network on pixels", "CNN from scratch",
           "transfer learning", "forest on hand-made columns, 150 images", "CNN from scratch, 150 images",
           "CNN from scratch, 150 images + augmentation", "transfer learning, 150 images"]).round(1)
results
```

Read the last four rows. On this data, across six measured random starts:
- the **scratch CNN on 150 images** drops and becomes unpredictable: anywhere from about 90% to 95%;
- the **same CNN with augmentation** is steady again, between about 94.5% and 96.5%: as good as with five times more labels;
- **transfer learning on 150 images** gives the same number every run: about 91% for this set of 150. Tag a different 150 and it lands between about 94% and 96%. With few labels, *which* images were tagged matters as much as the model;
- the **forest on the hand-made columns** needs no random start and holds about 94% on 150 images.

So, for hot spots: **keep the thermographer's columns as the model to beat**. With few labels, a pretrained network and a small CNN with the right augmentation reach the same level.
They are what you need when the defect cannot be written as a column. Most industrial vision systems today follow this pattern: a pretrained network, augmentation that respects the physics of the camera, and a few hundred images of your own.

Two kinds of mistakes:

- **Missed hot spots.** Most of the rest are hot spots called normal, some of them easy for you to see. The sun-heated cabinets, a weak spot of the thermographer's columns, are not where it fails.
  The pretrained network learned from everyday photos, and one new neuron on top can only do so much.

A common next step is *fine-tuning*: unfreeze the network's top layers and train them a little, with a very small step, on your images. It is not a sure win.
With a few hundred images from two substations, it can just as easily learn those two substations, so it must earn its place on validation like everything else.

### Where deep learning earns its cost

| Track | Deep learning? | Start from |
|---|---|---|
| B: thermal images | **Not for a plain hot spot**: the hottest pixel above the rest is a column an engineer can write, and here it is as good as a CNN. **Yes** when the defect is a shape nobody can write as a column (a cracked insulator, a missing bolt) | the hand-made columns + a forest as the model to beat; then a pretrained CNN + a new last layer + augmentation that never changes the colours |
| C: raw logger windows | Sometimes: when summaries (Lab 6) do not separate the loggers that lie | a 1-D CNN on 24-hour windows of loading |
| A: load forecast over time | Rarely at first: after a boosting baseline on lag columns | a sequence model, only if it beats the baseline |
| D: asset register of 900 transformers | Almost never | boosting or a forest (Lab 5) |

Deep learning is data hungry, compute hungry, harder to explain, and has more knobs. Use it where the columns must be learned from the raw signal.
Where an engineer can write the column, try that first: it is quicker to build and clearer to explain.

## What you learned
- A neuron is a weighted sum plus a bend (ReLU inside, sigmoid at the output for yes/no). Training is Lab 1's gradient step, repeated, with backpropagation computing the slopes.
- A convolution slides a small shared filter over the input (strictly a cross-correlation: no flip); stacked layers learn edges, then blobs, then parts, and you can look at the filters and their feature maps.
- A dense network on raw pixels fails: it does not know which pixels are neighbours. A CNN learns its columns from the pixels.
- When an engineer can write the column (the hottest pixel above the rest), a forest on it can be as good as a CNN: keep it as the model to beat.
- Transfer learning (a frozen pretrained network + one new last layer) is quick and steady, even on false-colour thermal images; augmentation of the training images (never validation, and never a change of colour in a thermal image) makes a small network steady with few labels.
- Split by substation, judge on the substation the model never saw, and open the mistakes, including the wrong labels: the score then stands clearly and honestly.
- Deep learning earns its cost where the columns cannot be written by hand; on a table of 900 rows, Lab 5 wins.
